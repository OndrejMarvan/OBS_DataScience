# Restaurants — Exploratory Data Analysis

**Project 1 (Classification)** — Machine Learning 1, WNE UW, 2025/2026

Predict whether a restaurant has closed (`status_closed`). Metric: **balanced accuracy**.

## Goals of this notebook

1. Understand the structure of the data (shapes, dtypes, missingness).
2. Inspect the **target distribution** — class imbalance is the central modeling challenge.
3. Identify which features are numeric / categorical / boolean.
4. Visualize key distributions and relationships with the target.
5. List the preprocessing decisions that will feed into `02_search.ipynb`.


## 1. Imports and data loading

```python
import warnings
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 200)
sns.set_theme(style="whitegrid")

DATA_DIR = Path("data")
RANDOM_STATE = 42
```

```python
train = pd.read_csv(DATA_DIR / "restaurants_train.csv")
test = pd.read_csv(DATA_DIR / "restaurants_test.csv")
sample_sub = pd.read_csv(DATA_DIR / "restaurant_sample_submission.csv")

print(f"Train:        {train.shape}")
print(f"Test:         {test.shape}")
print(f"Sample sub:   {sample_sub.shape}")
```

```python
train.head()
```

## 2. Target variable

The target is `status_closed`: 1 if the restaurant has closed, 0 if still open.

```python
target = "status_closed"
counts = train[target].value_counts()
proportions = train[target].value_counts(normalize=True)

print("Class counts:")
print(counts)
print("\nClass proportions:")
print(proportions.round(4))
```

```python
fig, ax = plt.subplots(1, 2, figsize=(11, 4))

counts.plot(kind="bar", ax=ax[0], color=["#2E5984", "#C44536"])
ax[0].set_title("Class counts")
ax[0].set_xlabel("status_closed")
ax[0].set_ylabel("count")
ax[0].set_xticklabels(["0 (open)", "1 (closed)"], rotation=0)

proportions.plot(kind="bar", ax=ax[1], color=["#2E5984", "#C44536"])
ax[1].set_title("Class proportions")
ax[1].set_xlabel("status_closed")
ax[1].set_ylabel("proportion")
ax[1].set_xticklabels(["0 (open)", "1 (closed)"], rotation=0)
ax[1].set_ylim(0, 1)

for axis in ax:
    for p in axis.patches:
        axis.annotate(f"{p.get_height():.0f}" if axis is ax[0] else f"{p.get_height():.3f}",
                      (p.get_x() + p.get_width() / 2, p.get_height()),
                      ha="center", va="bottom")

plt.tight_layout()
plt.show()
```

**Observation:** the dataset is severely imbalanced — roughly **90% open / 10% closed**. This drives several design choices for the modeling notebook:

- The grading metric is **balanced accuracy** (= (sensitivity + specificity) / 2), which is the appropriate metric for imbalanced binary classification. Plain accuracy would be misleading: a model predicting "open" for everyone would already score 0.90 on accuracy, but 0.50 on balanced accuracy.
- All cross-validation runs in `02_search.ipynb` will use `scoring="balanced_accuracy"`.
- We will use `class_weight="balanced"` in supported estimators, and consider SMOTE / undersampling via `imblearn.pipeline.Pipeline` if it improves CV scores.
- The CV splitter will be `StratifiedKFold` to preserve class proportions in each fold.

## 3. Column types

We classify the 85 features into three groups based on dtype: numeric, categorical (object/str), and boolean.

```python
from pandas.api.types import is_bool_dtype, is_numeric_dtype

# Drop the ID and target before classifying feature columns
feature_cols = [c for c in train.columns if c not in {"restaurant_id", "status_closed"}]

bool_cols = [c for c in feature_cols if is_bool_dtype(train[c])]
numeric_cols = [c for c in feature_cols if is_numeric_dtype(train[c]) and not is_bool_dtype(train[c])]
categorical_cols = [c for c in feature_cols if c not in bool_cols + numeric_cols]

print(f"Numeric features:      {len(numeric_cols)}")
print(f"Categorical features:  {len(categorical_cols)}")
print(f"Boolean features:      {len(bool_cols)}")
print(f"Total:                 {len(feature_cols)}")
```

```python
print("Categorical columns:")
for c in categorical_cols:
    print(f"  {c:35s}  nunique={train[c].nunique()}, sample values={train[c].dropna().unique()[:5].tolist()}")
```

`category_top20` is the main categorical column with 20 levels — small enough for one-hot encoding. The two yes/no flags (`weekends_only`, `workdays_only`) are stored as objects but represent boolean information; we'll treat them as binary indicators in preprocessing.

```python
print("Boolean columns:")
for c in bool_cols:
    pos_rate = train[c].mean()
    print(f"  {c:35s}  positive rate = {pos_rate:.3f}")
```

## 4. Missingness

Some columns have a lot of missing values. We need to decide how to impute (or whether to drop) each one.

```python
missing = train[feature_cols].isna().sum().sort_values(ascending=False)
missing_pct = (missing / len(train) * 100).round(2)
missing_table = pd.DataFrame({"missing_count": missing, "missing_pct": missing_pct})
missing_table = missing_table[missing_table["missing_count"] > 0]

print(f"Columns with any missing values: {len(missing_table)}")
missing_table.head(20)
```

```python
fig, ax = plt.subplots(figsize=(10, max(4, len(missing_table) * 0.25)))
missing_table["missing_pct"].plot(kind="barh", ax=ax, color="#C44536")
ax.invert_yaxis()
ax.set_xlabel("% missing")
ax.set_title("Missingness by column (train set)")
ax.axvline(50, color="black", linestyle="--", linewidth=0.8, label="50% threshold")
ax.legend()
plt.tight_layout()
plt.show()
```

**Observation:** missingness ranges from ~0% to over 70%. Notable cases:

- `tagcat_payment_options` (~72%) and `tagcat_social_inclusivity` (~38%) — these are sparse "tag category" features, missingness itself may be informative.
- `price_level` (~51%) — likely informative missingness (the establishment never disclosed its price level).
- `ratings_avg_*_prior` and `ratings_num_*_prior` — missing for newer restaurants without enough history.

**Strategy:** rather than dropping high-missingness columns (which can carry signal in the missingness pattern), we will impute with the **median** for numeric and the **most frequent** for categorical, and let regularization handle any noise. We may also add **missing indicators** (`SimpleImputer(add_indicator=True)`) as a low-cost feature engineering step.

## 5. Numeric feature distributions

Plot histograms for a sample of numeric features to see scale and skew. Many features are likely heavy-tailed (counts, ratings, distances). StandardScaler will handle this in the pipeline, but visualizing helps confirm.

```python
# Pick a representative subset of numeric features to plot
features_to_plot = [
    "user_ratings_total",
    "price_level",
    "hours_open",
    "rating_avg",
    "place_age_days",
    "review_length_avg",
    "ratings_num_12m_prior",
    "ratings_avg_12m_prior",
    "catch_restaurant_count_500m",
    "residents",
    "restaurants_per_capita",
    "poi_count_500m",
]

fig, axes = plt.subplots(4, 3, figsize=(14, 12))
for ax, col in zip(axes.flat, features_to_plot):
    train[col].hist(ax=ax, bins=50, color="#2E5984")
    ax.set_title(col, fontsize=10)
plt.tight_layout()
plt.show()
```

## 6. Numeric features vs target

For each selected feature, compare its distribution between open (0) and closed (1) restaurants. Features with clearly different distributions across classes are likely to be predictive.

```python
features_vs_target = [
    "user_ratings_total",
    "rating_avg",
    "place_age_days",
    "ratings_num_12m_prior",
    "hours_open",
    "review_length_avg",
]

fig, axes = plt.subplots(2, 3, figsize=(14, 8))
for ax, col in zip(axes.flat, features_vs_target):
    sns.boxplot(data=train, x=target, y=col, ax=ax, palette=["#2E5984", "#C44536"], showfliers=False)
    ax.set_title(f"{col} by status_closed")
    ax.set_xlabel("")
    ax.set_xticklabels(["open (0)", "closed (1)"])
plt.tight_layout()
plt.show()
```

**Reading the plots:**

- `user_ratings_total`, `ratings_num_12m_prior`: closed restaurants tend to have **fewer total/recent ratings** — engagement is a strong signal.
- `rating_avg`: closed restaurants tend to have slightly **lower average ratings**.
- `place_age_days`: closed restaurants skew **younger** — many closures happen in the first few years of operation. Confirms that age is informative.
- `hours_open`, `review_length_avg`: smaller class differences, may be weaker predictors but contribute jointly.

## 7. Categorical feature: `category_top20` vs target

Some restaurant categories may have higher closure rates than others.

```python
cat_summary = (
    train.groupby("category_top20", observed=True)[target]
    .agg(["count", "mean"])
    .rename(columns={"count": "n", "mean": "closure_rate"})
    .sort_values("closure_rate", ascending=False)
)
cat_summary.head(20)
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
cat_summary["closure_rate"].plot(kind="barh", ax=ax, color="#C44536")
ax.invert_yaxis()
ax.axvline(train[target].mean(), color="black", linestyle="--", linewidth=0.8, label="overall closure rate")
ax.set_xlabel("closure rate")
ax.set_title("Closure rate by restaurant category (category_top20)")
ax.legend()
plt.tight_layout()
plt.show()
```

Categories vary in closure rate, so the variable carries signal. With only 20 levels, **one-hot encoding** is the natural choice (no high-cardinality blowup).

## 8. Correlation between numeric features

Highly correlated features will be handled by **regularization** (elastic net) rather than manual removal — the course covers this approach in the regression notebook. Still, visualizing helps identify clusters of redundant features.

```python
# Use a subset of numeric features to keep the heatmap readable
numeric_subset = [c for c in numeric_cols if c not in ["restaurant_id"]][:30]
corr = train[numeric_subset].corr()

fig, ax = plt.subplots(figsize=(13, 11))
sns.heatmap(corr, ax=ax, cmap="RdBu_r", center=0, vmin=-1, vmax=1, square=True, cbar_kws={"shrink": 0.7}, linewidths=0.3)
ax.set_title("Correlation matrix — first 30 numeric features")
plt.tight_layout()
plt.show()
```

Visible clusters (e.g. the `ratings_num_*_prior` family, the `catch_*` neighborhood family, the `gus_*` socioeconomic family) confirm we should rely on regularization rather than manual feature removal — these blocks of correlated columns are exactly what elastic net is designed to handle.

## 9. Summary — preprocessing decisions for `02_search.ipynb`

| Step | Decision | Reason |
|---|---|---|
| **CV scoring** | `balanced_accuracy` | Matches the grading metric |
| **CV splitter** | `StratifiedKFold(n_splits=5, shuffle=True, random_state=42)` | Preserves 90/10 class ratio per fold |
| **Class imbalance** | `class_weight="balanced"` first; consider SMOTE / undersampling via `imblearn` Pipeline if it helps | Free, often best; resampling only if CV says so |
| **Numeric features** | `SimpleImputer(strategy="median")` + `StandardScaler` | Robust to outliers; scaling required by KNN / SVM / regularized logreg |
| **Categorical (`category_top20`)** | `SimpleImputer(strategy="most_frequent")` + `OneHotEncoder(handle_unknown="ignore")` | 20 levels is fine for one-hot |
| **Boolean features** | Cast to int (no scaling needed for 0/1) | They're already on a comparable scale |
| **Yes/No object features** (`weekends_only`, `workdays_only`) | Map to 0/1 first, then treat as boolean | Same logic |
| **High-missingness columns** | Keep, impute with median, no special dropping | Missingness pattern may carry signal; regularization handles noise |
| **Algorithms to compare** | Logistic Regression (Elastic Net) · KNN · SVM (RBF) | Required ≥ 3, in scope per ML1 syllabus |

**Risks flagged:**

- SVM with RBF kernel on 33k rows is O(n²). We'll benchmark on a 5k subsample first; fall back to `LinearSVC` if too slow.
- KNN may underperform on 85-dimensional data (curse of dimensionality), but we include it as the third required model.

Next step: `02_search.ipynb` builds the preprocessing `ColumnTransformer`, fits each algorithm with cross-validated hyperparameter search, and selects the best for the final submission.
