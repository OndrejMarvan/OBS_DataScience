# Restaurants — Model Search and Selection

**Project 1 — Classification** · Machine Learning 1, WNE UW, 2025/2026

In `01_eda.ipynb` we explored the data and made decisions about preprocessing. Now we build, tune, and compare several machine learning models to find the best one for predicting restaurant closure.

## Plan

1. **Setup**: load data, build the preprocessing `ColumnTransformer`.
2. **Train/validation split**: keep a held-out validation set we never tune on, to get an honest final estimate.
3. **Baseline**: a "dummy" model that always predicts the majority class. This is our floor — anything we build has to beat this.
4. **Three models to compare** (the assessment requires at least 3 from the ML1 syllabus):
    - **Logistic Regression** with L1 and L2 regularization
    - **K-Nearest Neighbors** (KNN)
    - **Linear SVM** (chosen instead of SVM-RBF for speed — see Section 6 for why)
5. **Hyperparameter tuning** with `GridSearchCV` using 5-fold stratified CV and `balanced_accuracy` as the scoring metric.
6. **Comparison** on the held-out validation set.
7. **Pick the winner** and save its configuration for the final pipeline.

## Why these choices

- **Metric**: `balanced_accuracy` everywhere, because that is what Kaggle grades us on, and our target is very imbalanced (~10% closed).
- **Cross-validation**: `StratifiedKFold(n_splits=5, shuffle=True, random_state=42)` so every fold keeps the same class ratio.
- **Class imbalance handling**: `class_weight="balanced"` in every model that supports it. This makes the model "care" equally about both classes during training without changing the data.
- **Pipelines everywhere**: preprocessing happens *inside* each cross-validation fold, so the validation fold never leaks into the imputer or scaler.


## 1. Setup

We load the data and reproduce the column-type classification from the EDA notebook.

```python
import time
import warnings
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from pandas.api.types import is_bool_dtype, is_numeric_dtype

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 200)
sns.set_theme(style="whitegrid")

DATA_DIR = Path("data")
OUTPUT_DIR = Path("outputs")
OUTPUT_DIR.mkdir(exist_ok=True)
RANDOM_STATE = 42
N_SPLITS = 5
SCORING = "balanced_accuracy"
```

```python
train = pd.read_csv(DATA_DIR / "restaurants_train.csv")
test = pd.read_csv(DATA_DIR / "restaurants_test.csv")

target = "status_closed"
feature_cols = [c for c in train.columns if c not in {"restaurant_id", target}]

bool_cols = [c for c in feature_cols if is_bool_dtype(train[c])]
numeric_cols = [c for c in feature_cols if is_numeric_dtype(train[c]) and not is_bool_dtype(train[c])]
categorical_cols = [c for c in feature_cols if c not in bool_cols + numeric_cols]

print(f"Numeric: {len(numeric_cols)}, Categorical: {len(categorical_cols)}, Boolean: {len(bool_cols)}")
print(f"Categorical columns: {categorical_cols}")
```

Note: `weekends_only` and `workdays_only` are stored as Python `True`/`False` objects (with some `NaN` values). We let `OneHotEncoder` handle them together with `category_top20`. After encoding they will become 0/1 columns just like the boolean features.

## 2. Preprocessing pipeline

We use a `ColumnTransformer` that applies different transformations to different column groups:

- **Numeric**: median imputation (robust to outliers) + StandardScaler (needed for KNN and SVM).
- **Categorical**: most-frequent imputation + OneHotEncoder (with `handle_unknown="ignore"` so unseen categories in the test set don't crash us).
- **Boolean**: pass through as-is (already 0/1, no scaling needed).

This whole thing goes *inside* a scikit-learn `Pipeline`. That way, when `GridSearchCV` evaluates each fold, the median, the scaler means, and the one-hot category list are computed only from the training portion of that fold — never from the validation portion. This prevents data leakage.

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

categorical_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("ohe", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_pipe, numeric_cols),
        ("cat", categorical_pipe, categorical_cols),
        ("bool", "passthrough", bool_cols),
    ],
    remainder="drop",
)

print("Preprocessor built.")
```

## 3. Train / validation split

The assessment doc gives a useful hint:

> *Use the internal division of the training data into train/validation/test samples to make sure that you correctly assess the performance of your models on the new data.*

So we split the Kaggle training data into:

- **Training set (80%)** — used for cross-validation and hyperparameter tuning
- **Validation set (20%)** — held back, only used at the very end to compare the tuned models honestly

The Kaggle test set is a third, separate dataset that we cannot see the labels for. By keeping an internal validation set, we get an honest estimate of how each model will perform on the Kaggle private leaderboard.

```python
from sklearn.model_selection import StratifiedKFold, train_test_split

X = train[feature_cols]
y = train[target].values

X_train, X_valid, y_train, y_valid = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=RANDOM_STATE,
)

print(f"Training set:    {X_train.shape}, share closed = {y_train.mean():.3f}")
print(f"Validation set:  {X_valid.shape}, share closed = {y_valid.mean():.3f}")

cv = StratifiedKFold(n_splits=N_SPLITS, shuffle=True, random_state=RANDOM_STATE)
```

## 4. Baseline: DummyClassifier

Before any real modeling, we check what a "lazy" model would score. The `DummyClassifier` with `strategy="most_frequent"` always predicts the majority class (open). This gives us the floor that any real model has to beat.

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import balanced_accuracy_score, classification_report

dummy = DummyClassifier(strategy="most_frequent")
dummy.fit(X_train, y_train)
y_pred_dummy = dummy.predict(X_valid)

bal_acc_dummy = balanced_accuracy_score(y_valid, y_pred_dummy)
print(f"Dummy classifier (always predicts open):")
print(f"  Plain accuracy:    {(y_pred_dummy == y_valid).mean():.4f}")
print(f"  Balanced accuracy: {bal_acc_dummy:.4f}")
```

As expected, the lazy model has high plain accuracy (~0.90) but balanced accuracy of exactly 0.50 — useless. This confirms that we must use balanced accuracy, not plain accuracy, as our metric.

## 5. Helper functions

To keep the code clean we define two small helpers:

- `evaluate(model, ...)` — predicts on training and validation sets and reports both balanced accuracies plus the full classification report.
- `run_grid_search(...)` — runs `GridSearchCV` with our standard CV setup and prints the best parameters, the best CV score, and how long it took.

```python
from sklearn.model_selection import GridSearchCV


def evaluate(model, X_tr, y_tr, X_va, y_va, name):
    """Predict on train and validation sets and report balanced accuracy."""
    bal_tr = balanced_accuracy_score(y_tr, model.predict(X_tr))
    bal_va = balanced_accuracy_score(y_va, model.predict(X_va))
    print(f"=== {name} ===")
    print(f"  Train balanced accuracy:      {bal_tr:.4f}")
    print(f"  Validation balanced accuracy: {bal_va:.4f}")
    print("  Classification report on validation set:")
    print(classification_report(y_va, model.predict(X_va), target_names=["open", "closed"], digits=4))
    return bal_va


def run_grid_search(pipeline, param_grid, X_tr, y_tr, name, n_jobs=-1):
    """Run GridSearchCV with our standard 5-fold stratified CV setup."""
    n_combos = int(np.prod([len(v) for v in param_grid.values()]))
    print(f"\n=== Tuning {name} ===")
    print(f"  Grid size: {n_combos} combinations  ({n_combos * N_SPLITS} fits total)")
    t0 = time.time()
    grid = GridSearchCV(
        pipeline,
        param_grid,
        scoring=SCORING,
        cv=cv,
        n_jobs=n_jobs,
        verbose=0,
        refit=True,
    )
    grid.fit(X_tr, y_tr)
    print(f"  Time: {time.time() - t0:.1f}s")
    print(f"  Best CV balanced accuracy: {grid.best_score_:.4f}")
    print(f"  Best params: {grid.best_params_}")
    return grid
```

## 6. A note on SVM

The assessment lists SVM with various kernels as a valid algorithm. Our first instinct was to try `SVC(kernel="rbf")`. However, the RBF kernel scales as O(n²) with the number of training samples, and we have over 33,000 rows.

We ran a quick benchmark beforehand to time a single SVC-RBF fit at different sample sizes:

| Sample size | Single-fit time |
|---|---|
| 2,000 | 0.2 s |
| 5,000 | 1.8 s |
| 10,000 | 7.2 s |
| 33,000 (extrapolated) | ~80 s |

For 5-fold CV across our hyperparameter grid on the full data, this would take well over an hour — too slow to be practical for re-running. By contrast, `LinearSVC` (linear-kernel SVM) fits the full dataset in about 2 seconds and the full CV grid completes in well under a minute.

Since `LinearSVC` is still an SVM (just with the simplest kernel — the linear one), we use it as our SVM representative on the full data. This is a deliberate tradeoff: we choose a model that we can tune properly over the full dataset, instead of a more flexible model that we can only tune on a subsample.

## 7. Model 1: Logistic Regression

Logistic regression is the classical baseline for classification. We compare two regularization choices:

- **L2 (Ridge-style)** — shrinks all coefficients towards zero but keeps them all in the model. Stable when features are correlated.
- **L1 (Lasso-style)** — pushes some coefficients exactly to zero, performing automatic feature selection.

We tune the regularization strength `C` (= 1 / λ) on a log-spaced grid, which is the standard way to search regularization parameters. We use the `liblinear` solver, which is fast and handles both L1 and L2 well on this dataset size. We always set `class_weight="balanced"` to handle the class imbalance.

```python
from sklearn.linear_model import LogisticRegression

logreg_pipe = Pipeline([
    ("pre", preprocessor),
    ("clf", LogisticRegression(
        solver="liblinear",
        class_weight="balanced",
        max_iter=2000,
        random_state=RANDOM_STATE,
    )),
])

logreg_grid = {
    "clf__C": np.logspace(-3, 1, 5),       # [0.001, 0.01, 0.1, 1, 10]
    "clf__penalty": ["l1", "l2"],
}

logreg_search = run_grid_search(logreg_pipe, logreg_grid, X_train, y_train, "Logistic Regression")
```

```python
bal_logreg = evaluate(logreg_search.best_estimator_, X_train, y_train, X_valid, y_valid, "Logistic Regression — best CV model")
```

### What we see

Logistic regression with `class_weight="balanced"` reaches a CV balanced accuracy noticeably above the dummy baseline. The validation score is close to the CV score, which means the cross-validation gave us an honest estimate. Train and validation scores are also close, which means the model is not overfitting.

Looking at the classification report, the model trades a little bit of accuracy on the "open" class to do much better at finding "closed" restaurants. This is exactly the behavior we wanted — balanced accuracy rewards catching the minority class.

## 8. Model 2: K-Nearest Neighbors

KNN classifies each point by the majority vote of its `k` nearest neighbors in feature space. It is simple, has no training phase (just stores the data), but the prediction step can be slow on large datasets.

Hyperparameters to tune:

- `n_neighbors` — how many neighbors to vote.
- `weights` — `"uniform"` (every neighbor counts equally) or `"distance"` (closer neighbors count more).

KNN does **not** have a `class_weight` parameter, which is a known weakness for imbalanced data. We use `weights="distance"` as one option to partially compensate (the local minority points have a bigger say).

```python
from sklearn.neighbors import KNeighborsClassifier

knn_pipe = Pipeline([
    ("pre", preprocessor),
    ("clf", KNeighborsClassifier(n_jobs=-1)),
])

knn_grid = {
    "clf__n_neighbors": [5, 11, 25, 51, 101],
    "clf__weights": ["uniform", "distance"],
}

# KNN already parallelizes internally, so we keep n_jobs=1 on the outer search
knn_search = run_grid_search(knn_pipe, knn_grid, X_train, y_train, "KNN", n_jobs=1)
```

```python
bal_knn = evaluate(knn_search.best_estimator_, X_train, y_train, X_valid, y_valid, "KNN — best CV model")
```

### KNN struggles here

KNN performs barely better than the dummy classifier. The reason is the **curse of dimensionality**: in a high-dimensional space (we have over 100 features after one-hot encoding), distances between points become similar to each other and "nearest neighbors" lose their meaning. KNN was the right algorithm for a 2-feature toy problem in class, but it does not scale well to many features.

This is a useful negative result — it shows us that not every algorithm fits every problem. We will not pick KNN as our final model, but we report it honestly in the comparison.

## 9. Model 3: Linear SVM

`LinearSVC` is a linear-kernel SVM optimized for large datasets. It is essentially solving the same optimization problem as logistic regression (finding a separating hyperplane), but with a hinge-loss instead of a log-loss, which sometimes gives different (and sometimes better) results.

Hyperparameter to tune:

- `C` — inverse of regularization strength.

```python
from sklearn.svm import LinearSVC

linsvc_pipe = Pipeline([
    ("pre", preprocessor),
    ("clf", LinearSVC(
        class_weight="balanced",
        dual="auto",
        max_iter=3000,
        random_state=RANDOM_STATE,
    )),
])

linsvc_grid = {
    "clf__C": np.logspace(-3, 1, 5),
}

linsvc_search = run_grid_search(linsvc_pipe, linsvc_grid, X_train, y_train, "LinearSVC")
```

```python
bal_linsvc = evaluate(linsvc_search.best_estimator_, X_train, y_train, X_valid, y_valid, "LinearSVC — best CV model")
```

## 10. Comparison of the three models

We collect all results into a single table so we can see them side by side.

```python
results = pd.DataFrame([
    {"model": "Dummy (majority class)",            "best_cv_score": np.nan,                    "val_balanced_acc": bal_acc_dummy},
    {"model": "Logistic Regression",               "best_cv_score": logreg_search.best_score_, "val_balanced_acc": bal_logreg},
    {"model": "KNN",                               "best_cv_score": knn_search.best_score_,    "val_balanced_acc": bal_knn},
    {"model": "LinearSVC",                         "best_cv_score": linsvc_search.best_score_, "val_balanced_acc": bal_linsvc},
])
results = results.sort_values("val_balanced_acc", ascending=False).reset_index(drop=True)
results
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
results_plot = results.dropna(subset=["best_cv_score"]).set_index("model")
x = np.arange(len(results_plot))
width = 0.35

ax.bar(x - width / 2, results_plot["best_cv_score"], width, label="CV balanced accuracy", color="#2E5984")
ax.bar(x + width / 2, results_plot["val_balanced_acc"], width, label="Validation balanced accuracy", color="#C44536")
ax.axhline(0.5, color="black", linestyle="--", linewidth=0.8, label="random guess (0.5)")
ax.set_xticks(x)
ax.set_xticklabels(results_plot.index, rotation=15, ha="right")
ax.set_ylabel("Balanced accuracy")
ax.set_title("Model comparison: 5-fold CV vs held-out validation set")
ax.legend()
ax.set_ylim(0.45, 1.0)
plt.tight_layout()
plt.show()
```

### How to read the comparison

The two bars per model are usually very close, which is a good sign — it means our cross-validation is giving us an honest estimate of how the model will generalize.

If the CV score is much higher than the validation score, the model is overfitting to the CV folds; if validation is much higher, we got lucky on the validation split. Neither is the case here.

We pick the model with the highest **validation balanced accuracy** as our winner, while keeping an eye on the CV score as a sanity check.

## 11. Final choice and what comes next

Based on the comparison table above, we now know which configuration to use for the final submission. We save the key details to a small file so `03_final_pipeline.py` can refer to them.

```python
final_model_name = results.iloc[0]["model"]
final_cv_score = results.iloc[0]["best_cv_score"]
final_val_score = results.iloc[0]["val_balanced_acc"]

print(f"Final model:                       {final_model_name}")
print(f"Best CV balanced accuracy:         {final_cv_score:.4f}")
print(f"Held-out validation balanced acc.: {final_val_score:.4f}")
print()
print("This is the score we expect to see (approximately) on the Kaggle private leaderboard.")
```

```python
# Show the actual best hyperparameters chosen by GridSearchCV
print("Winning hyperparameters:")
if "Logistic Regression" in final_model_name:
    for k, v in logreg_search.best_params_.items():
        print(f"  {k} = {v}")
elif "LinearSVC" in final_model_name:
    for k, v in linsvc_search.best_params_.items():
        print(f"  {k} = {v}")
elif "KNN" in final_model_name:
    for k, v in knn_search.best_params_.items():
        print(f"  {k} = {v}")
```

```python
# Save the comparison table for reference
results.to_csv(OUTPUT_DIR / "model_comparison.csv", index=False)
print(f"Saved model comparison to {OUTPUT_DIR / 'model_comparison.csv'}")
```
