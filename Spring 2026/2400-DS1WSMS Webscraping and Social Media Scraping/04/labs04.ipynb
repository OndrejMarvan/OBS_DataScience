```python
import requests # The "Courier" that goes to the website and grabs the code
from lxml import html # The "Organizer" that turns messy HTML into a neat tree

# STEP 1: Set the Target
# This is the website we want to "read"
url = "https://www.wne.uw.edu.pl/"

# STEP 2: The Request
# We tell the computer: "Go to this URL and bring back the raw HTML text"
response = requests.get(url)

# STEP 3: Create the "Tree"
# We turn that giant wall of text into an 'lxml tree' so we can use our XPath
tree = html.fromstring(response.content)
tree 
# <Element html at 0x123abc>. That’s just Python’s way of saying, "I have the map in my memory, but I’m waiting for you to tell me what to look at."
```

# Use a Relative XPath to grab specific text. If you want to see all the news titles on the WNE page:

```python
# STEP 4: Use the XPath "Address" from the Plugin
# This is the path you copied! 
# IMPORTANT: I added '/text()' at the end to get the words inside the box.
# my_xpath = '/html/body/div[1]/div[2]/main/div[2]/div/div[1]/div[1]/div/div[1]/h1/text()'
my_xpath = '//h1//text()'

# We tell the tree: "Follow this path and tell me what you find"
data = tree.xpath(my_xpath)

# STEP 5: Show ALL Results
if data:
    print(f"I found {len(data)} items:")
    
    # We use a 'for loop' to go through the list one by one
    for item in data:
        clean = item.strip()
        if clean: # Only print if it's not just an empty space
            print(f" -> {clean}")
else:
    print("Nothing found!")
```

### Example: Scrapping agendas of DS&BA programm courses offered by the University of Warsaw

Before we start: can we scrap this content?

USOS does not have robots.txt file. Also, there are no terms of use/service. No problem then?

However, the courses' agendas can be perceived as an intellectural property (notably, not owned by the lecturers, rather by the University). Still, the University shares them openly, i.e. everyone can access them, without loggin into USOS. Can one have some problems scrapping them? After all, we aim to collect them for purely educational, private, eventually scientific purposes.

Obviously, on no account should we break the website or terribly slow it down as it would possibly constitute a crime. 

Conclusion: we can scrap it, yet not sure if for commercial purposes (intellectual property).

Step 1: Accessing certain USOS webpage html structure; initial indentification of interesting parts.

```python
url = "https://usosweb.wne.uw.edu.pl/kontroler.php?_action=katalog2%2Fprzedmioty%2FszukajPrzedmiotu&_prz_kod_pattern=2400-DS&lang=en"

webpage_html = requests.get(url) # request for a webpage structure
# print(webpage_html.text) # .text provides html only
```

Step 2: Collecting links to subpages with agendas.

```python
url = "https://usosweb.wne.uw.edu.pl/kontroler.php?_action=katalog2%2Fprzedmioty%2FszukajPrzedmiotu&_prz_kod_pattern=2400-DS&lang=en"

webpage_html = requests.get(url) # request for a webpage structure

soup = BeautifulSoup(webpage_html.text, 'html.parser') # one need to format it into a BeautifulSoup object before proceeding

hrefs = []
for a in soup.find_all('a', class_='wrblue'): # finds all elements 'a' with attribute 'class'='wrblue'
    hrefs.append(a['href'])
        
print(hrefs)
```

Step 3: Building a scraper handling a single course agenda.

```python
url = hrefs[0] # let us consider only the first website for a while

response = requests.get(url) # loading the website
soup = BeautifulSoup(response.text, 'html.parser') # parsing website's content

# here, we find all the tables;
# please pay attention to the difference between `find()` and `find_all()`
# `find()` returns the first occurence of an element; `find_all()` obviously returns all of them as a list
tables = soup.find_all('table')

# let us take the tables, one by one, and convert them from html content to pandas.DataFrame
tables_df = dict()
for i in range(0,len(tables)):
    table_html = str(tables[i]) # iterating through tables
    html_io = StringIO(table_html) # wrapping HTML in StringIO to avoid FutureWarning
    tables_df[i] = pd.read_html(html_io)[0] # reading tables using pandas
```

```python
len(tables_df) # looks like there are 4 tables
```

```python
tables_df[1] # we look for this table, i.e. the one with agenda
```

```python
# this reverses the table above; the first column are the colnames, the second column are the values to be included in a row
new_df = pd.DataFrame([tables_df[1][1].values], columns=tables_df[1][0].values)
new_df
```

Step 4: Scaling the scraper, i.e. collecting all agendas in a loop.

```python
start = time.time()

url = "https://usosweb.wne.uw.edu.pl/kontroler.php?_action=katalog2%2Fprzedmioty%2FszukajPrzedmiotu&_prz_kod_pattern=2400-DS&lang=en"

webpage_html = requests.get(url) # request for a webpage structure

soup = BeautifulSoup(webpage_html.text, 'html.parser') # one need to format it into a BeautifulSoup object before proceeding

hrefs = []
for a in soup.find_all('a', class_='wrblue'): # finds all elements 'a' with attribute 'class'='wrblue'
    hrefs.append(a['href'])

new_dfs = dict(); index = 0
for url in hrefs:

    response = requests.get(url) # loading the website
    soup = BeautifulSoup(response.text, 'html.parser') # parsing website's content

    tables = soup.find_all('table') # here, we find all the tables
    
    table_html = str(tables[1]) # let us take the second table each time, looks like it will include all the content
    html_io = StringIO(table_html) # wrapping HTML in StringIO to avoid FutureWarning
    table_df = pd.read_html(html_io)[0] # reading tables using pandas

    # this reverses the table above; the first column are the colnames, the second column are the values to be included in a row
    new_df = pd.DataFrame([table_df[1].values], columns=table_df[0].values)
    new_dfs[index] = new_df; index += 1

end = time.time()
print(end - start)
```

```python
result = pd.concat(new_dfs.values(), axis=0) # let us concat all the agendas collected
result.head()
```

```python
# here, we save what we scraped as pickle or txt
# obviously we could use some different formats though
save_object(result, r'outputs/USOS_courses_agendas.pkl')
save_object(result, r'outputs/USOS_courses_agendas.txt')
```

### Example: scraping U.S. Supreme Court website for the decisions of the Court

Before we start: can we scrap this content?

Let us examine the robots.txt file, see: https://www.supremecourt.gov/robots.txt. Clearly we can scrap the content of this website but the administrator asks us to use a delay between requests of 1 second. Probably our scraper will be blocked if we ignore it.

On the website there are no other information regarding possible webscraping of its content. Specifically there are no terms of use/service due to the fact that this is a public institution's website.

It is hard to imagine us breaking the law 'in general' by scraping this website (no intelectual property or personal data here, no damage caused by a scraper), unless we break the website or terribly slow it down - then it can be considered a crime.

Conclusion: we can scrap it, but let us use 1 s. delay consistently with robots.txt.

Step 1: Accessing U.S. Supreme Court's webpage html structure; initial indentification of interesting parts.

```python
url = "https://www.supremecourt.gov/opinions/slipopinion/24" # looks like the number at the end denotes year, good for us!

webpage_html = requests.get(url) # request for a webpage structure
#print(webpage_html.text) # .text provides html only
```

Step 2: Collecting all links reffering to subpages.

```python
url = "https://www.supremecourt.gov/opinions/slipopinion/24"

webpage_html = requests.get(url) # request for a webpage structure

soup = BeautifulSoup(webpage_html.text, 'html.parser') # one need to format it into a BeautifulSoup object before proceeding

hrefs = []
for a in soup.find_all('a'): # finds all elements 'a'
    try:
        hrefs.append(a.attrs['href']) # finds all elements 'href'
    except:
        continue
        
# print(hrefs)
```

Step 3: Looking for links including U.S. Supreme Court opinions

```python
url = "https://www.supremecourt.gov/opinions/slipopinion/24"

webpage_html = requests.get(url) # request for a webpage structure

soup = BeautifulSoup(webpage_html.text, 'html.parser') # one need to format it into a BeautifulSoup object before proceeding

hrefs = []
for a in soup.find_all('a'): # finds all elements 'a'
    try:
        hrefs.append(a.attrs['href']) # finds all elements 'href'
    except:
        continue
        
pdfs = []
for href in hrefs:
    if('opinions/24') in href: # takes only those that include phrase 'opinons/23', i.e. are opinions from 2023
        pdfs.append(href)
        
pdfs
```

```python
# what we have collected is not enough, we need to add 'https://www.supremecourt.gov' at the beggining of each address
# it is not the most efficient way to do it, but let us stick to lambda function discussed during this course
pdfs_list = list(pd.Series(pdfs).map(lambda x: "https://www.supremecourt.gov"+x))
pdfs_list
```

Step 4: accessing the opinions and reading them

```python
# !pip3 install requests PyPDF2
```

```python
import requests
import PyPDF2
import io  # <--- CRITICAL: You need this to handle the "Bytes" from the internet

# 1. Get the PDF from the web
response = requests.get(pdfs_list[0]) 

# 2. Convert the raw bytes into a format PyPDF2 can read
# Think of io.BytesIO as a "virtual file" that lives in your RAM
pdf_memory_file = io.BytesIO(response.content)
pdf_reader = PyPDF2.PdfReader(pdf_memory_file)

text = ""

# 3. Extract text from every page
# Pro-tip: New PyPDF2 allows you to loop directly over 'pdf_reader.pages'
for page in pdf_reader.pages:
    text += page.extract_text() + "\n" # Adding a newline keeps words from sticking together

print(f"Successfully extracted {len(text)} characters from the PDF.")
print(text)
```

Step 5: scraping all U.S. Supreme Court opinions from 2023-2024

CAUTION! It appears we do not have enough time to scrap everything. Let us collect first 30 texts.

```python
start = time.time()

pdfs = []

for year in range(24, 16, -1): # year = {2023, 2022, ..., 2017}
    
    url = "https://www.supremecourt.gov/opinions/slipopinion/"+str(year) # we will go through different years

    time.sleep(1) # 1 s. delay as we were asked for

    webpage_html = requests.get(url) # gets the website
    soup = BeautifulSoup(webpage_html.text, 'html.parser') # takes the HTML structure

    hrefs = []
    for a in soup.find_all('a'): # finds all elements 'a'
        try:
            hrefs.append(a.attrs['href']) # finds all elements 'href'
        except:
            continue
        
    for href in hrefs:
        if('opinions/'+str(year)) in href: # within 'href's finds those that include a pattern 'opinions/'
            pdfs.append(href)
        
pdfs_list = list(pd.Series(pdfs).map(lambda x: "https://www.supremecourt.gov"+x)) # adding the prefix to all links

texts = []

for pdf in pdfs_list[0:30]: # for each pdf address; here we limit to 30

    time.sleep(1) # 1 s. delay as we were asked for

    try:

        response = requests.get(pdf) # get the structure of the PDF
        
        pdf_file = PyPDF2.PdfReader(io.BytesIO(response.content)) # convert bytes response to output with text

        text = ""
        for page_num in range(len(pdf_file.pages)): # for each page in the PDF...
            text += pdf_file.pages[page_num].extract_text() # ...get text
            
        texts.append(text) # append to list of texts

    except:
        continue

end = time.time()
print(end-start)
```

```python
print(len(texts)) # number of collected texts
#texts[0] # an individual document
```

```python
# here, we save what we scraped as pickle or txt
# obviously we could use some different formats though
save_object(texts, r'outputs/output_USsupremecourt.pkl')
save_object(texts, r'outputs/output_USsupremecourt.txt')
```
