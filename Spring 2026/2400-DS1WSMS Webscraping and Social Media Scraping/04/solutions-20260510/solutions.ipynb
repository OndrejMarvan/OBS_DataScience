##### Exercise 1. 

From the webpage https://en.wikipedia.org/wiki/List_of_hard_rock_bands_(N%E2%80%93Z) 

extract links to web pages containing information about musicians whose name begin with ”Q”. 

From all the collected links extract: the name of the band, the genre, number of years active. Present the collected data in a pandas DataFrame.

```python
import requests
from lxml import html
import pandas as pd
import time
import io

# STEP 1: Set up the "Identity" of your scraper
# This tells Wikipedia who you are and why you are there
headers = {
    "User-Agent": "UN-Flags-Scraper/1.0 (student project; contact: your_email@example.com)"
}

main_url = "https://en.wikipedia.org/wiki/List_of_hard_rock_bands_(N%E2%80%93Z)"

# We pass the 'headers' into the requests.get() function
response = requests.get(main_url, headers=headers)
tree = html.fromstring(response.content)

# Target 'Q' section links
q_band_links = tree.xpath('/html/body/div[3]/div/div[3]/main/div[3]/div[3]/div[2]/div[9]')
# Instead of looking for ID="Q", we look for the header that CONTAINS "Q"
q_band_links = tree.xpath('//h2[contains(., "Q")]/following-sibling::ul[1]/li/a/@href')

base_url = "https://en.wikipedia.org"
data_list = []

print(f"Requesting with User-Agent: {headers['User-Agent']}")
print(f"Found {len(q_band_links)} 'Q' bands. Starting extraction...")

for link in q_band_links:
    full_url = base_url + link
    
    # Use headers for every single page request!
    res = requests.get(full_url, headers=headers)
    band_tree = html.fromstring(res.content)
    
    try:
        # Extracting data using the "Infobox" strategy
        name = band_tree.xpath('//h1/i/text() | //h1/text()')[0]
        genre = band_tree.xpath('normalize-space(//th[text()="Genres"]/following-sibling::td)')
        years = band_tree.xpath('normalize-space(//th[text()="Years active"]/following-sibling::td)')
        
        data_list.append({
            "Band Name": name,
            "Genre": genre,
            "Years Active": years
        })
        print(f"Scraped: {name}")
    except Exception:
        print(f"Skipping: {link} (Incomplete data)")
    
    time.sleep(0.2) # Polite delay

df = pd.DataFrame(data_list)
print("\nDone!")
print(df.head())
```

##### Exercise 2. 

From the webpage https://en.wikipedia.org/wiki/Lists_of_musicians 

extract links to webpages containing information about musicians specialising in music genre beginning with "A". 

From all the links obtained in the previous step, extract links to artists' webpages. 

From the obtained artists' webpages extract: name of the band, and years active. 

As an output use default print function on pandas data frame.


```python
import requests
from lxml import html
import pandas as pd
import re
import time

# STEP 1: Identification
headers = {
    "User-Agent": "UN-Flags-Scraper/1.0 (student project; contact: your_email@example.com)"
}
base_url = "https://en.wikipedia.org"

# STEP 2: Level 1 - The Master List
main_url = "https://en.wikipedia.org/wiki/Lists_of_musicians"
response = requests.get(main_url, headers=headers)
tree = html.fromstring(response.content)

# Generalized Path for all list links
all_links = tree.xpath('//main//div/ul/li/a/@href')

# REGEX FILTER: Find genre lists that start with "A"
 # When we used r"^/wiki/List_of_[aA]", we were looking for pages that start exactly with "List_of_A...". 
# # However, the A Cappella page is plural: /wiki/Lists_of_a_cappella_groups.
genre_regex = r"^/wiki/List_of_[aA]"
a_genre_links = [l for l in all_links if re.match(genre_regex, l, re.I)]

print(f"Level 1: Found {len(a_genre_links)} genre pages starting with 'A'.")

# --- MODIFIED STEP 3: Tracking the Source ---
# We use a dictionary to keep track of: {artist_link: genre_link}
artist_to_genre_map = {}

for g_link in a_genre_links[:3]: # Sample of 3 genres
    print(f" -> Accessing Genre: {g_link}")
    res_g = requests.get(base_url + g_link, headers=headers)
    tree_g = html.fromstring(res_g.content)
    
    g_page_links = tree_g.xpath('//main//div/ul/li/a[1]/@href')
   
    
    for l in g_page_links:
        if "/wiki/" in l and ":" not in l:
            # We save the artist link as the KEY and the genre as the VALUE
            if l not in artist_to_genre_map:
                artist_to_genre_map[l] = g_link
    
    time.sleep(0.1)

print(f"Level 2: Collected {len(artist_to_genre_map)} unique artist URLs.")

# STEP 4: Level 3 - Deep Dive into Artist Pages
final_results = []

# Scrape a sample of 10 artists
for a_link, source_genre in list(artist_to_genre_map.items()):
    try:
        res_a = requests.get(base_url + a_link, headers=headers)
        tree_a = html.fromstring(res_a.content)
        
        # 1. Extract name and years active
        name = tree_a.xpath('normalize-space(//h1)')
        years = tree_a.xpath('normalize-space(//th[contains(., "Years active")]/following-sibling::td)')
        
        # 2. CLEANING THE GENRE TEXT
        # This removes '/wiki/', 'List_of_', and turns underscores into spaces
        clean_genre = source_genre.replace("/wiki/", "").replace("List_of_", "").replace("Lists_of_", "").replace("_", " ")
        
        # 3. ADDING COLUMNS
        final_results.append({
            "Band Name": name,
            "Years Active": years if years else "N/A",
            "Genre Name": clean_genre,  # Cleaned text version
            # "Source Link": base_url + source_genre # Original URL for reference
        })
        print(f"    Scraped: {name} from {clean_genre}")
    except:
        continue
    time.sleep(0.1)

# STEP 5: Final Output
df = pd.DataFrame(final_results)
print("\n--- FINAL PANDAS DATAFRAME ---")
print(df)
```
