````python
import requests
from bs4 import BeautifulSoup

# 1. Download the page
url = "http://pythonscraping.com/pages/warandpeace.html"
html = requests.get(url).text
soup = BeautifulSoup(html, "html.parser")

# 2️. Get all <p> tags (paragraphs)
ps = soup.findAll("p")
# ps = soup.select("div#text p")
# 3️. Print how many paragraphs we found
print("Number of paragraphs:", len(ps))
non_empty = [p for p in ps if p.get_text(strip=True)]
print(ps)
# # 4️. Print only the first 2 paragraph texts
for p in non_empty:
    print(p.get_text())
````

#### Mini-Exercise 4: Extract All Links from a Website

In this exercise, you will practice:

- Sending an HTTP request to a real website  
- Parsing HTML using BeautifulSoup  
- Extracting all hyperlinks (`<a>` tags)  
- Accessing both visible text and link URLs  

We will use the website: https://www.wne.uw.edu.pl  

You are given a website. Your job is to:

##### 1. Download the webpage
- Use `requests.get()` to retrieve the HTML.
- Store the HTML content.
- Create a `BeautifulSoup` object.


##### 2. Extract all `<a>` tags
- Use `soup.findAll("a")` to find all hyperlinks.
- Count how many links are on the page.
- Print the total number.

##### 3. Print link text and URL
For every `<a>` tag:

- Extract:
  - The visible text (what users see on the page)
  - The `href` attribute (the actual link)
- If `href` is missing, print `"NO HREF"`
- Display the result in this format:


````python
import requests
from bs4 import BeautifulSoup

# 1: Download the webpage

# Send HTTP request to the website
url = "https://www.wne.uw.edu.pl"
html = requests.get(url).text   # Retrieve HTML content

# Create BeautifulSoup object (parse HTML)
soup = BeautifulSoup(html, "html.parser")

# Task 2: Extract all <a> tags

# Find all hyperlink tags
links = soup.findAll("a")

# Count total number of links
print("Total links found:", len(links))
print("-" * 40)

# Task 3: Print link text and URL

for a in links:
    # Extract visible text (what users see)
    text = a.get_text().strip()
    
    # Extract href attribute (actual URL)
    href = a.get("href")
    
    # If href does not exist, replace with "NO HREF"
    if href is None:
        href = "NO HREF"
    
    # Display in required format
    print(text, "->", href)
````

##### Mini-Exercise 5: Extract and Download Image URLs from a Website
In this exercise, you will practice:
Sending an HTTP request to a website
Parsing HTML using BeautifulSoup
Extracting image elements (<img> tags)
Converting relative URLs into absolute URLs
Removing duplicate URLs
Creating a folder if it does not exist
Downloading and saving images locally
We will use the website:
https://www.wne.uw.edu.pl/
##### Your Task
You are given a website. Your job is to extract image information and download the images locally.
##### 1. Download and Parse the Webpage
Use requests.get() to retrieve the HTML (add timeout).
Create a BeautifulSoup object to parse the page.
##### 2. Extract All <img> Tags
Use soup.find_all("img") to find all image elements.
Count how many <img> tags exist on the page.
Print the total number.
##### 3. Extract and Clean Image URLs
For each image:
Get the src attribute (image source).
Skip images with missing src.
Convert relative URLs into absolute URLs using urljoin().
Remove duplicate image URLs (keep order).
Print the first 15 image URLs.
##### 4. Create a Folder for Images
Create a folder (e.g., downloaded_images).
If the folder does not exist, create it using os.makedirs().
##### 5. Download and Save Images
For each cleaned image URL:
Send a request to download the image.
Extract the filename from the URL.
Save the image inside the created folder.
Handle possible download errors.

Tip:
Images are embedded in HTML like this:

- How images are embedded in HTML:
  ```html
  <img src="/images/logo.png">





````python
import os
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse

url = "https://www.wne.uw.edu.pl/"
html = requests.get(url, timeout=20).text
soup = BeautifulSoup(html, "html.parser")

# 1. Extract all <img> tags

imgs = soup.find_all("img")
print("IMG tags:", len(imgs))

image_urls = []
for img in imgs:
    src = img.get("src")
    if not src:  # skip if src missing
        continue
    image_urls.append(urljoin(url, src))

# 2. Remove duplicates (keep order)
seen = set()
image_urls = [u for u in image_urls if not (u in seen or seen.add(u))]

print("\nFirst 15 image URLs:")
for u in image_urls[:15]:
    print(u)

# 3. Create folder (if not exists)

folder_name = "downloaded_images_wne"

if not os.path.exists(folder_name):
    os.makedirs(folder_name)

# 4. Download images

for img_url in image_urls:
    try:
        response = requests.get(img_url, timeout=20)
        if response.status_code == 200:
            
            # Extract filename from URL
            parsed_url = urlparse(img_url)
            filename = os.path.basename(parsed_url.path)
            
            # If filename is empty, skip
            if not filename:
                continue
            
            file_path = os.path.join(folder_name, filename)
            
            # Save image
            with open(file_path, "wb") as f:
                f.write(response.content)
                
            print("Saved:", filename)
            
    except Exception as e:
        print("Failed:", img_url)
````

````python
import os
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse

url = "https://www.wne.uw.edu.pl/"
html = requests.get(url, timeout=20).text
soup = BeautifulSoup(html, "html.parser")

os.makedirs("wne_images", exist_ok=True)

imgs = soup.find_all("img")
image_urls = []

for img in imgs:
    src = img.get("src")
    if not src:
        continue
    image_urls.append(urljoin(url, src))

# unique
seen = set()
image_urls = [u for u in image_urls if not (u in seen or seen.add(u))]

download_n = 10
count = 0

for img_url in image_urls:
    if count >= download_n:
        break

    try:
        r = requests.get(img_url, timeout=20)
        if r.status_code != 200:
            continue

        # make a filename
        path = urlparse(img_url).path
        filename = os.path.basename(path) or f"image_{count}.jpg"
        outpath = os.path.join("wne_images", filename)

        with open(outpath, "wb") as f:
            f.write(r.content)

        print("Saved:", outpath)
        count += 1

    except Exception as e:
        print("Skipped:", img_url, "|", e)

print("Downloaded:", count)
````
