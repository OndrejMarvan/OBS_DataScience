### How to extract that table ? find_all()

````python
tables = soup.find_all('table')
tables
````

````python
tables = soup.find_all('table')
# List to store DataFrames
dfs = []
for table in tables:
    # Extract data from the table and create a DataFrame
    rows = []
    for row in table.find_all('tr')[1:]:  # Skip the header row
        columns = row.find_all('td')
        # print(columns)
        rows.append([col.get_text(strip=True) for col in columns])

    # Define column names
    columns = ['Item Title', 'Description', 'Cost', 'Image']
    

    # Create a Pandas DataFrame
    df = pd.DataFrame(rows, columns=columns)

    # Append the DataFrame to the list
    dfs.append(df)


dataset_1 = pd.DataFrame(dfs[0])

# table.find_all('img')

dataset_1
````

#### What does that mean?

#### 1. What is a User-Agent?

A **User-Agent** is like your **ID card on the internet**.

When you visit a website using Chrome, your browser automatically says something like:

> “Hi, I’m Chrome browser version XX on Windows.”

But when your Python script makes a request without a User-Agent, it basically says:

> “Hi, I’m… nothing.”

That looks suspicious 
So Wikipedia may block or warn your request.

#### 2. What is a robot policy?

Websites have **rules for bots (robots)**.

Wikipedia’s robot policy says:

✔ Identify yourself with a User-Agent  
✔ Don’t send too many requests too fast  
✔ Be respectful (add delay between requests)

It’s like **knocking politely on the door instead of breaking into the house**.

````python
import requests
from bs4 import BeautifulSoup
import time

url = "https://pt.wikipedia.org/wiki/Lista_de_bairros_de_Manaus"

headers = {
    "User-Agent": "MyResearchBot/1.0 (your_email@example.com)"
}

# Be polite — small delay before request
time.sleep(1)

response = requests.get(url, headers=headers)

if response.status_code == 200:
    soup = BeautifulSoup(response.text, "html.parser")
    print(soup.title.text)
else:
    print("Failed to retrieve page:", response.status_code)
````

#### What is `urllib.robotparser`?

`urllib.robotparser` is a Python tool that reads a website's **robots.txt** file.

A `robots.txt` file tells bots:

- Which pages they are allowed to visit
- Which pages they are NOT allowed to visit

It is basically the website’s **rulebook for bots**.

#### What does this code do?

```python
import urllib.robotparser

rp = urllib.robotparser.RobotFileParser()
rp.set_url("https://pt.wikipedia.org/robots.txt")
rp.read()

print(rp.can_fetch("*", url))

````python
import urllib.robotparser

rp = urllib.robotparser.RobotFileParser()
rp.set_url("https://pt.wikipedia.org/robots.txt")
rp.read()

print(rp.can_fetch("*", url))
````

If the answer is False, it means the page is disallowed for automated scraping.

Why would Wikipedia block it?
Possible reasons:
The page is protected from automated scraping
The site wants you to use their official API instead
The request looks like an anonymous bot
Some parts of Wikipedia are restricted for bots

````python
import requests
from bs4 import BeautifulSoup

api_url = "https://pt.wikipedia.org/w/api.php"
headers = {"User-Agent": "MyResearchBot/1.0 (your_email@example.com)"}

params = {
    "action": "parse",
    "page": "Lista_de_bairros_de_Manaus",
    "prop": "text",
    "format": "json",
    "formatversion": 2,
}

r = requests.get(api_url, params=params, headers=headers, timeout=30)

ct = r.headers.get("Content-Type", "")
print("status:", r.status_code)
print("content-type:", ct)

if r.status_code != 200:
    print("Not OK. First 500 chars:\n", r.text[:500])
elif "application/json" not in ct.lower():
    print("Not JSON. First 500 chars:\n", r.text[:500])
else:
    data = r.json()
    html = data["parse"]["text"]
    soup = BeautifulSoup(html, "html.parser")
    print("Parsed OK. Page title:", soup.get_text(" ", strip=True)[:120])
````

````python
import requests
from bs4 import BeautifulSoup
import time

url = "https://pt.wikipedia.org/wiki/Lista_de_bairros_de_Manaus"

headers = {
    "User-Agent": "MyResearchBot/1.0 (your_email@example.com)"
}

# Be polite
time.sleep(1)

response = requests.get(url, headers=headers, timeout=30)

if response.status_code == 200:
    soup = BeautifulSoup(response.text, "html.parser")

    # Get all tables
    tables = soup.find_all("table")
    print(f"Total tables found: {len(tables)}")

    # Find wikitable sortable
    table = soup.find("table", class_="wikitable sortable")

    if table:
        print("Target table found ✅")
    else:
        print("Target table NOT found ❌")

else:
    print("Failed:", response.status_code)
````

Your goal is to build a small book scraper step by step.

🅐 A) Connect to the Website

🅑 B) Extract Book Titles (Page 1)

````python
# 🅐 A) Connect to the Website
# - Send a request to the homepage.
# - Create a BeautifulSoup object.
# - Print the page `<title>`.

import requests
from bs4 import BeautifulSoup

url = "http://books.toscrape.com/"

# Optional but good practice: add User-Agent
headers = {
    "User-Agent": "StudentScraper/1.0 (student@example.com)"
}

# Send request
response = requests.get(url, headers=headers)

# Check if request was successful
if response.status_code == 200:
    # Create BeautifulSoup object
    soup = BeautifulSoup(response.text, "html.parser")
    
    # Print page title
    print(soup.title.text)
else:
    print("Failed to retrieve page:", response.status_code)
````

````python
# 🅑 B) Extract Book Titles (Page 1)
# - Scrape all 20 book titles from the first page.
# - Store them in a list.
# - Print them numbered from 1 to 20.

import requests
from bs4 import BeautifulSoup

url = "http://books.toscrape.com/"

headers = {
    "User-Agent": "StudentScraper/1.0 (student@example.com)"
}

response = requests.get(url, headers=headers)

if response.status_code == 200:
    soup = BeautifulSoup(response.text, "html.parser")
    
    # Find all book containers
    books = soup.find_all("article", class_="product_pod")
    
    # Extract titles
    titles = []
    for book in books:
        title = book.h3.a["title"]
        titles.append(title)
    
    # Print numbered list
    for i, title in enumerate(titles, start=1):
        print(f"{i}. {title}")
        
else:
    print("Failed to retrieve page:", response.status_code)

````

````python
# 🅒 C) Extract Titles + Prices
# - For each book on page 1, collect:
#   - title
#   - price
# - Store results as a list of dictionaries:
# {"title": "...", "price": "..."}

import requests
from bs4 import BeautifulSoup

url = "http://books.toscrape.com/"

headers = {
    "User-Agent": "StudentScraper/1.0 (student@example.com)"
}

response = requests.get(url, headers=headers)

if response.status_code == 200:
    soup = BeautifulSoup(response.text, "html.parser")
    
    books = soup.find_all("article", class_="product_pod")
    
    books_data = []
    
    for book in books:
        title = book.h3.a["title"]
        price = book.find("p", class_="price_color").text
        
        books_data.append({
            "title": title,
            "price": price
        })
    
    # Print results
    for item in books_data:
        print(item)

else:
    print("Failed to retrieve page:", response.status_code)
````
