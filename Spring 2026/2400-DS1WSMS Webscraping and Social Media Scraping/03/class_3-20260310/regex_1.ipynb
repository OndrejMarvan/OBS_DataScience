### Python Regex Metacharacters

The real power of regex matching in Python emerges when <regex> contains **special characters called metacharacters.** These have a unique meaning to the regex matching engine and vastly enhance the capability of the search.

Consider again the problem of how to determine **whether a string contains any three consecutive decimal digit characters.**

In a regex, a **set of characters specified in square brackets ([]) makes up a character class.** 

This metacharacter sequence matches any single character that is in the class, as demonstrated in the following example:

````python
s = 'foo123bar'
re.search('[0-9][0-9][0-9]', s) 

# [0-9] matches any single decimal digit character—any character between '0' and '9', inclusive. 
# The full expression [0-9][0-9][0-9] matches any sequence of three decimal digit characters. 
# In this case, s matches because it contains three consecutive decimal digit 
````

````python
# These strings also match: So we do not care what are those numbers
print(re.search('[0-9][0-9][0-9]', 'foo456bar'))
print(re.search('[0-9][0-9][0-9]', '234baz'))
print(re.search('[0-9][0-9][0-9]', 'qux678'))
````

#### We may use different regex to extract same thing 

````python
# Search for three digits in a row using regex
re.search('\d\d\d', s)  
# This works, but Python shows a warning:
# "invalid escape sequence '\d'"
# because '\d' is not a valid escape in normal Python strings.
# Python interprets '\' as the start of an escape sequence.
````

````python
# Correct way: use a raw string so '\' is treated literally
re.search(r'\d\d\d', s)  
# r'' tells Python not to interpret backslashes as escape characters.
# \d means "any digit", so \d\d\d matches three digits in a row.
````

````python
s
````

````python
# Another equivalent regex using a character class
re.search('[0-9]+', s)  
# [0-9] means any digit from 0 to 9
# + means "one or more occurrences"
````

````python
# Search for exactly three digits in a row
re.search('[0-9]{3}', s)  
# [0-9] means any digit from 0 to 9
# {3} means the preceding pattern must appear exactly three times
# So the pattern matches exactly three digits in a row (e.g., 123, 456)

# This is equivalent to writing:
re.search(r'\d\d\d', s)

# A shorter regex version is:
re.search(r'\d{3}', s)
# \d means any digit
# {3} means exactly three repetitions
````

````python
# Search for three consecutive digits in a string
print(re.search('[0-9][0-9][0-9]', '12foo34'))

# [0-9] means any digit (0–9)
# The pattern [0-9][0-9][0-9] requires three digits in a row

# In the string '12foo34' there are digits,
# but they are not three consecutive digits

# '12' → only two digits
# '34' → only two digits

# Therefore, the regex does not find a match
# Output: None
````

````python
# [0-9] matches any digit character (0–9)

# The pattern [0-9][0-9] means:
# match two digits in a row

# In the string 'foo123bar', the first sequence of two digits is '12'

print(re.search('[0-9][0-9]', 'foo123bar'))

# Output:
# <re.Match object; span=(3, 5), match='12'>
# span=(3,5) means the match starts at position 3 and ends at position 5
# match='12' is the first sequence of two digits found in the string
````

### 1. **The dot (.) metacharacter** matches any character except a newline, so it functions like a wildcard:

* the regex 1.3 matches '123' because the '1' and '3' match literally, and the . matches the '2'. 
* Here, you’re essentially asking, “Does s contain a '1', then any character (except a newline), then a '3'?” 
* The answer is yes for 'foo123bar' but no for 'foo13bar'.

````python
s = 'foo1F3bar'
print(re.search('1.3', s))

# The dot (.) is a metacharacter that matches any single character
# Pattern: 1.3 means:
# 1  → the digit 1
# .  → any character
# 3  → the digit 3

# In 'foo123bar', the substring '123' matches:
# 1 matches '1'
# . matches '2'
# 3 matches '3'
````

````python
s = 'foo13bar'
print(re.search('1.3', s))

# In this string there is no character between '1' and '3'
# The pattern requires exactly one character between them
# Therefore, no match is found

# Output: None
````

### 2. Square brackets ([]) represent a character class

Characters contained in **square brackets ([]) represent a character class** —an enumerated set of characters to match from. 

A character class metacharacter sequence will match any single character contained in the class.

The metacharacter sequence [artz] matches any single 'a', 'r', 't', or 'z' character. In the example, the regex ba[artz] matches both **'bar' and 'baz'** (and would also match 'baa' and 'bat').

````python
print(re.search('ba[artz]', 'foobarqux'))
# [artz] is a character class
# It means the last character can be: a, r, t, or z
# 'bar' matches because r is in the set
# Output: match 'bar'


print(re.search('ba[artz]', 'foobazqux'))
# 'baz' matches because z is in the character class [artz]
# Output: match 'baz'


print(re.search('ba[artz]', 'foobatqux'))
# 'bat' matches because t is in the character class [artz]
# Output: match 'bat'


print(re.search('ba[artz]', 'foobaaqux'))
# 'baa' matches because a is also included in [artz]
# Output: match 'baa'
````

A character class can also contain a range of characters separated by a hyphen (-), in which case it matches any single character within the range. 

**For example, [a-z] matches any lowercase alphabetic character between 'a' and 'z', inclusive:*

````python
print(re.search('[a-z]', 'FOObar'))
# [a-z] matches any lowercase letter (a–z)

# In the string 'FOObar', the first lowercase letter is 'b'
# Therefore the regex finds a match: 'b'


print(re.search('[A-Z]', 'FOObar'))
# [A-Z] matches any uppercase letter (A–Z)

# In the string 'FOObar', the first uppercase letter is 'F'
# Therefore the regex finds a match: 'F'
````

````python
print(re.search('[a-z]', 'FOO'))
````

### 3.  ^ - Caret The caret symbol ^ is used to check if a string <font color='red'>starts</font>    with a certain character.

````python
# Example list of animal names
mylist = ["dog", "cat", "wildcat", "thundercat", "cow", "hooo", "rabbit", "raccoon"]

# 1. Print animals starting with "r"
# ^ is an anchor that means "start of the string"
# ^r means the word must start with the letter r

r = re.compile("^r")
newlist =list(filter(r.match, mylist))  # keep only elements that match the pattern
newlist
````

````python
# Example list of animal names
mylist = ["dog", "cat", "wildcat", "thundercat", "cow", "hooo", "rabbit", "raccoon"]

# 1. Print animals starting with "r"
# ^ is an anchor that means "start of the string"
# ^r means the word must start with the letter r

r = re.compile("^r")
newlist = list(filter(r.match, mylist))  # keep only elements that match the pattern
print(newlist)

# Output:
# ['rabbit', 'raccoon']


# 2. Print animals starting with "ra"
# ^ra means the word must start with the letters "ra"

r = re.compile("^ra")
newlist = list(filter(r.match, mylist))
print(newlist)

# Output:
# ['rabbit', 'raccoon']


# 3. Print animals starting with "rac"
# ^rac means the word must start with "rac"

r = re.compile("^rac")
newlist = list(filter(r.search, mylist))
print(newlist)

# Output:
# ['raccoon']
````

### 4. $ - Dollar The dollar symbol is used to check if a string <font color='green'>ends</font>  with a certain character.

````python
# Example list of animal names
mylist = ["dog", "cat", "wildcat", "thundercat", "cow", "hooo", "rabbit", "raccoon"]

# 1. Print animals that end with the letter "t"
# $ is an anchor meaning "end of the string"
# t$ means the word must end with the letter "t"

r = re.compile("t$")
newlist = list(filter(r.search, mylist))  # keep only elements that match the pattern
print(newlist)

# Output:
# ['cat', 'wildcat', 'thundercat', 'rabbit']


# 2. Print animals that end with "it"
# it$ means the word must end with the letters "it"

r = re.compile("it$")
newlist = list(filter(r.search, mylist))
print(newlist)

# Output:
# ['rabbit']
````

### 5. * - Star The star symbol * matches <font color='#8a0f8a'>zero or more occurrences of the pattern left to it</font> .

````python
# Example list of animal names
mylist = ["dog", "cat", "wildcat", "thundercat", "cow", "hooo", "rabbit", "raccoon"]

# 1. Print animals that contain "oo" inside the word
# .* means "any characters (zero or more)"
# oo is the sequence we are searching for
# .*oo means: any characters followed by "oo"

r = re.compile(".*oo")
newlist = list(filter(r.search, mylist))  # keep elements that match the pattern
print(newlist)

# Output:
# ['hooo', 'raccoon']


# 2. Print animals that end with "cat"
# .* means any characters before
# cat is the ending we are searching for
# .*cat means: any characters followed by "cat"

mylist = ["dog", "cat", "wildcat", "thundercat", "cow", "hooo"]
r = re.compile(".*cat")
newlist = list(filter(r.search, mylist))
print(newlist)

# Output:
# ['cat', 'wildcat', 'thundercat']
````

### 6. + - Plus The plus symbol + matches one or more occurrences of the pattern left to it.

````python
# Example list of animal names
mylist = ["dog", "cat", "wildcat", "thundercat", "cow", "hooo", "rabbit", "raccoon"]

# Print animals that contain "co", "coo", "cooo", etc.
# + means "one or more occurrences of the preceding character"

# co+ means:
# c followed by one or more "o"

r = re.compile("co+")
newlist = list(filter(r.search, mylist))  # keep elements that match the pattern
print(newlist)

# Output:
# ['cow', 'raccoon']
````

### 7. ? - Question Mark The question mark symbol ? matches zero or one occurrence of the pattern left to it.

````python
# Example list of words
mylist = ["mn", "man", "maaan", "main", "woman"]

# ? means "zero or one occurrence of the preceding character"
# ma?n means:
# m followed by optional "a", followed by n

# So it matches:
# mn  → a appears 0 times
# man → a appears 1 time

r = re.compile("ma?n")
newlist = list(filter(r.search, mylist))  # keep elements that match the pattern
print(newlist)

# Output:
# ['mn', 'man']
````

### 9. | - Alternation Vertical bar | is used for alternation (or operator).

````python
# Example list of animal names
mylist = ["dog", "cat", "wildcat", "thundercat", "cow", "hooo", "rabbit", "raccoon"]

# | means "OR" in regular expressions

# Pattern: a|h
# This means:
# match words that contain either the letter "a" OR the letter "h"

r = re.compile("a|h")

# filter keeps only elements where the regex pattern is found
newlist = list(filter(r.search, mylist))

print(newlist)

# Output:
# ['cat', 'wildcat', 'thundercat', 'hooo', 'rabbit', 'raccoon']
````

````python
# Example list of animal names
mylist = ["dog", "cat", "wildcat", "thundercat", "cow", "hooo", "rabbit", "raccoon"]

# ^ means "start of the string"
# | means "OR"

# Pattern: ^r|^c
# This means:
# words that start with "r" OR words that start with "c"

r = re.compile("^r|^c")

# filter keeps only elements where the regex pattern is found
newlist = list(filter(r.search, mylist))

print(newlist)

# Output:
# ['cat', 'cow', 'rabbit', 'raccoon']
````

### 10. () - Group Parentheses () is used to group sub-patterns. For example, (a|b|c)xz match any string that matches either a or b or c followed by xz

### 11. \ - Backslash Backlash \ is used to escape various characters including all metacharacters. 

For example, \$a match if a string contains dollar followed by a. Here, dollar is not interpreted by a RegEx engine in a special way.

````python
import re

# -------------------------------------------------------
# 1. Match strings that start with the literal characters "^r"
# and are followed by either "ab" or "ac"
# -------------------------------------------------------

mylist = ["dog", "$cat", "wildcat", "thundercat", "$cow", "hooo", "^rabbit", "^raccoon", "rabbit", "raccoon"]

r = re.compile(r"^\^r(ab|ac)")

# Explanation of the regex:
# ^      → start of the string
# \^     → literal caret character ^
# r      → the letter r
# (ab|ac) → followed by "ab" or "ac"

newlist = list(filter(r.search, mylist))
print(newlist)

# Output:
# ['^rabbit', '^raccoon']
````

````python
# -------------------------------------------------------
# 2. Why is this empty?
# -------------------------------------------------------

mylist = ["dog", "cat", "wildcat", "thundercat", "cow", "hooo", "^rabbit", "^raccoon"]

r = re.compile(r"^\r(ab|ac)")

# Explanation:
# ^   → start of the string
# \r  → carriage return character (special control character)

# This regex is looking for:
# start of string + carriage return + "ab" or "ac"

# None of the strings start with a carriage return,
# so no match is found.

newlist = list(filter(r.search, mylist))
print(newlist)

# Output:
# []


# Important rule for students:
# \^ → literal ^
# \$ → literal $
# \r → carriage return (not "^r")
````

````python
# -------------------------------------------------------
# 3. Match strings that contain literal "$" after t or w
# -------------------------------------------------------

mylist = ["dog", "cat", "wildcat$", "thundercat", "cow$", "hooo", "^rabbit", "^raccoon"]

r = re.compile(r"(t|w)\$")

# Explanation:
# (t|w) → letter t OR w
# \$    → literal dollar sign $

# So the pattern matches:
# "t$" or "w$"

newlist = list(filter(r.search, mylist))
print(newlist)

# Output:
# ['wildcat$', 'cow$']
````

````python
# -------------------------------------------------------
# 4. Match strings that END with "t" or "g"
# -------------------------------------------------------

mylist = ["dog", "cat", "wildcat$", "thundercat", "cow$", "hooo", "^rabbit", "^raccoon"]

r = re.compile(r"(t|g)$")

# Explanation:
# (t|g) → letter t OR g
# $     → end of the string

# So the word must end with "t" or "g"

newlist = list(filter(r.search, mylist))
print(newlist)

# Output:
# ['dog', 'cat', 'thundercat', '^rabbit']

# Why "^rabbit" matches:
# because the word ends with "t"
# the "^" at the beginning is just a normal character in the string
````

````python
import re

# -------------------------------------------------------
# Match metacharacters inside a character class
# -------------------------------------------------------

mylist = ["4[\\^\\.-", "+*/]\\d"]

# Pattern: [\^-]

r = re.compile(r"[\^-]")

# Explanation:
# [ ] → character class (match ONE character from the set)
#
# \^ → literal caret character ^
# -  → literal dash character -
#
# So the regex matches either:
# ^  OR  -

newlist = list(filter(r.search, mylist))

print(newlist)

# Output:
# ['4[\\^\\.-']
````

### 1. positive lookahead with ?= This is used to find a match only if it is succeeded by another pattern. 

````python
import re

# Example text containing names and gender tags
text = "Amanda:Female,Gerrard:Male,Lindsay:Female,Gianne:Female,Mike:Male"

# Regex pattern
re.findall(r"([A-Z]\w+):(?=Female)", text)

# Explanation of the regex:
#
# ([A-Z]\w+)   → capture a name
#                [A-Z]  = first letter must be uppercase
#                \w+    = one or more word characters (letters) [a-z]
#
# :            → the name must be followed by a colon
#
# (?=Female)   → positive lookahead
#                this means "Female must appear next",
#                but it is NOT included in the match

# Result:
# ['Amanda', 'Lindsay', 'Gianne']
````

````python
import re

# Example 1
print(re.search(r'geeks(?=[a-z])', "geeksforwebscrapping"))

# Explanation:
# geeks       → match the word "geeks"
# (?=[a-z])   → positive lookahead
#               check that the next character is a lowercase letter (a–z)

# In "geeksforwebscrapping":
# "geeks" is followed by "f"
# "f" is a lowercase letter → condition satisfied

# Output:
# <re.Match object ... match='geeks'>


````

````python
# Example 2
print(re.search(r'geeks(?=[a-z])', "webscrappingforgeeks"))

# In this string:
# "geeks" appears at the end of the string
# there is NO character after it

# The lookahead (?=[a-z]) requires a lowercase letter AFTER "geeks"
# Since there is nothing after "geeks", the condition fails

# Output:
# None
````

### 2. ?! Negative lookahead

This matches any pattern that is not succeeded by a specified pattern.

````python
import re

# In this example we have a string containing names and gender labels.
# Each entry follows this format:
# FirstName:Gender

text = "Amanda:Female,Gerrard:Male,Lindsay:Female,Gianne:Female,Mike:Male"


# Our goal:
# Extract the FIRST NAMES that are NOT followed by the word "Female".
#
# In other words, we want to find names that correspond to "Male".
#
# To do this we use a regex concept called a **negative lookahead**.


# Regex explanation:
#
# ([A-Z]\w+)
#   - Match a name that starts with a capital letter
#   - [A-Z] means the first character must be uppercase
#   - \w+ means one or more letters after it
#
# :
#   - Match the colon separating name and gender
#
# (?!Female)
#   - This is a NEGATIVE LOOKAHEAD
#   - It means: the text after the colon must NOT be "Female"
#
# So this pattern finds names whose label is not "Female".

result = re.findall(r"([A-Z]\w+):(?!Female)", text)

print(result)
````

### 3. ?<= Lookbehind assertion

This is the opposite of positive lookahead. 
In this case, we want to **match a pattern only if it is preceded by a specified pattern.** 

````python
import re

# In this example we have a string that contains three groups:
# fruits, animals, and colors.
#
# Each group follows the format:
# category={item1,item2,item3}

text = "fruits={apple,banana,orange} animals={dog,cat,rabbit} colors={blue,yellow,green}"


# Our goal:
# Extract ONLY the set of colors.
#
# We do this by matching the set that appears after the word "colors".
#
# To achieve this, we use a regex concept called a **positive lookbehind**.


# Regex explanation:
#
# (?<=colors)
#   - This is a POSITIVE LOOKBEHIND
#   - It means the match must be preceded by the word "colors"
#
# =
#   - Match the equals sign that appears after "colors"
#
# ({[\w,]+})
#   - Match the set of values inside curly braces
#   - { and } match the literal curly braces
#   - \w matches letters (word characters)
#   - , allows commas between items
#   - + means one or more characters
#
# The parentheses () create a capture group so only the set is returned.

result = re.findall(r"(?<=colors)=({[\w,]+})", text)

print(result)
````

````python
import re

# In this example we demonstrate **positive lookbehind**.
# A positive lookbehind checks that a certain pattern appears
# BEFORE the match, but it does not include it in the result.

# ------------------------------------------------
# Example 1
# ------------------------------------------------
# Regex: (?<=[a-z])\d{2}
#
# (?<=[a-z])
#   Positive lookbehind.
#   It requires that the characters before the match
#   contain a lowercase letter (a–z).
#
# \d{2}
#   Match exactly two digits.

print(re.search(r'(?<=[a-z])\d{2}', "geeks12"))

# Explanation:
# In the string "geeks12", the digits "12" come right after the letter "s",
# which is a lowercase letter. Therefore the condition is satisfied,
# and the match "12" is returned.



# ------------------------------------------------
# Example 2
# ------------------------------------------------
# Regex: (?<=geeks)\d{2}
#
# (?<=geeks)
#   Positive lookbehind that requires the word "geeks"
#   to appear immediately before the match.
#
# \d{2}
#   Match exactly two digits.

print(re.search(r'(?<=geeks)\d{2}', "geeks12"))

# Explanation:
# The digits "12" appear immediately after the word "geeks",
# so the pattern matches successfully and returns "12".
````

### 4.  ?<! Negative lookbehind

````python
import re

# In this example we again have three groups in the text:
# fruits, animals, and colors.
#
# Each group follows the format:
# category={item1,item2,item3}

text = "fruits={apple,banana,orange} animals={dog,cat,rabbit} colors={blue,yellow,green}"


# Our goal:
# Extract all sets that are NOT preceded by the word "colors".
#
# To do this we use a regex concept called a **negative lookbehind**.


# Regex explanation:
#
# (?<!colors)
#   - This is a NEGATIVE LOOKBEHIND.
#   - It means the text before the match must NOT be the word "colors".
#
# =
#   - Match the equals sign after the category name.
#
# ({[\w,]+})
#   - Capture the items inside the curly braces.
#   - { and } match the literal curly braces.
#   - \w matches letters (word characters).
#   - , allows commas between items.
#   - + means one or more characters.
#
# Parentheses () create a capture group so that only the set is returned.

print(re.findall(r"(?<!colors)=({[\w,]+})", text))
````

### 1. Extract information from txt files


In this exercise, you will practice using **regular expressions (regex)** to extract specific types of information from text files.

You are given a Python script that reads multiple text files from a folder and searches for different patterns using regular expressions.


Your goal is to understand how regex can be used to automatically extract structured information from unstructured text.

The script processes all `.txt` files located in the **`data`** folder and extracts the following information:

#### 1. Email addresses

Examples:
- `john@gmail.com`
- `user.name+tag@company.org`

#### 2. Age values

Examples:
- `Age: 41`
- `age 34`
- `29 years old`
- `Current age = 39`

#### 3. Income values

Examples:
- `$48,500`
- `USD 72,000`
- `83,900 dollars`
- `€51.700`

#### 4. Gene names

Examples:
- `BRCA1`
- `TP53`
- `EGFR`
- `KRAS`

The program reads each file, applies the regex patterns, and prints the extracted information.


#### What the Program Does

For every text file in the `data` folder, the program:

1. Reads the entire text file.
2. Uses regular expressions to search for:
   - email addresses
   - ages
   - incomes
   - gene names
3. Prints the extracted results for each file.

Example output:


````python
# import the regex library
# this lets us search text using patterns
import re

# Path helps us work with folders and files
from pathlib import Path


# this is the folder where our text files are stored
# example: data/file1.txt, data/file2.txt
folder = Path("data")

# -----------------------------
# REGEX PATTERNS
# -----------------------------

# this pattern tries to find email addresses
# example matches:
# john@gmail.com
# user.name+tag@company.org
# [A-Za-z0-9._%+\-]+    username
# @                     separator
# [A-Za-z0-9.\-]+       domain
# \.                    dot
# [A-Za-z]{2,}          tld
email_pattern = re.compile(
    r'[A-Za-z0-9._%+\-]+@[A-Za-z0-9.\-]+\.[A-Za-z]{2,}'
)


# this pattern tries to find ages in text
# examples it can match:
# age 34
# Age: 41
# Current age = 39
# 29 years old
age_pattern = re.compile(
    r'\b(?:age(?: now)?|Age|years|years old|Current age)\s*[:=]?\s*(\d{1,3})\b'
)


# this pattern tries to find income values
# examples it might match:
# $48,500
# USD 72,000
# 83,900 dollars
# €51.700
income_pattern = re.compile(
    r'(?:\$|USD\s?|EUR\s?|CAD\s?|£|€)?\s?(\d{1,3}(?:[,_\.]\d{3})*(?:\.\d+)?)\s?(?:USD|EUR|CAD|dollars)?',
    re.IGNORECASE
)


# this pattern finds gene names
# we list all genes we want to detect
# the regex will match any of them
gene_pattern = re.compile(
    r'\b(?:BRCA1|BRCA2|TP53|EGFR|KRAS|APOE(?:-e4)?|MTHFR|PALB2|ATM|CHEK2|PTEN|ALK|ROS1|MET|FTO|MC4R|CFTR|RB1|JAK2|CALR|MPL|HLA-B27|CYP2D6|COMT|BRIP1|RAD51C|RAD51D|APC|MLH1|MSH2|BRAF|NRAS)\b'
)


# -----------------------------
# PROCESS FILES
# -----------------------------

# loop through every text file in the folder
# example: file1.txt, file2.txt, file3.txt
for file_path in sorted(folder.glob("*.txt")):

    # read the entire text file into a string
    text = file_path.read_text(encoding="utf-8")


    # use regex to find all matches in the text
    # findall() returns a list of results
    emails = email_pattern.findall(text)
    ages = age_pattern.findall(text)
    incomes = income_pattern.findall(text)
    genes = gene_pattern.findall(text)


    # -----------------------------
    # PRINT RESULTS
    # -----------------------------

    # print a separator so output is easier to read
    print("=" * 60)

    # show which file we are processing
    print(f"FILE: {file_path.name}")

    # print all extracted information
    print("- Emails:", emails)
    print("- Ages:", ages)
    print("- Incomes:", incomes)
    print("- Genes:", genes)

    # blank line for readability
    print()
````

#### Web Scraping: Extracting Links with BeautifulSoup and Regex

In this exercise we will download a webpage and extract all hyperlinks (`<a>` tags) using two different approaches:

1. **BeautifulSoup (HTML parser)**
2. **Regular Expressions (Regex)**

The goal is to compare how both methods work and see whether they return the same results.

#### Step 1: Download the Webpage

We first send an HTTP request to the website using the `requests` library.  
The HTML content of the webpage is then stored in a variable called `html`.

```python
response = requests.get(url)
html = response.text
```

This HTML text contains the full structure of the webpage, including tags such as `<html>`, `<head>`, `<body>`, `<a>`, etc.


#### Step 2: Extract Links with BeautifulSoup

`BeautifulSoup` is a library designed specifically for parsing HTML and XML documents.

We create a parsed version of the HTML:

```python
soup = BeautifulSoup(html, "html.parser")
```

Then we find all `<a>` tags that contain an `href` attribute:

```python
soup.find_all("a", href=True)
```

From these tags we extract only the value of the `href` attribute.

Because some links may be **relative paths** (for example `/about` instead of `https://site.com/about`), we convert them into full URLs using:

```python
urljoin(base_url, link)
```


#### Step 3: Extract Links with Regex

The second method uses **regular expressions** to search directly inside the HTML text.

Regex pattern used:

```
<a[^>]*href="([^"]*)"
```

Explanation:

- `<a` matches the start of an anchor tag
- `[^>]*` matches any characters until the end of the tag
- `href="..."` captures the link inside quotation marks
- `([^"]*)` extracts the value of the link

We then use:

```python
re.findall(pattern, html)
```

to return all matches.

Just like before, we convert relative URLs into full URLs using `urljoin()`.

#### Step 4: Compare the Results

Both methods produce a list of links.  
To compare them easily, we store them in a **Pandas DataFrame**.

The table shows:

- Links extracted with BeautifulSoup
- Links extracted with Regex
- A boolean column (`Same?`) indicating whether both methods returned the same link

This allows us to observe differences between the two approaches.

#### Why Compare These Methods?

BeautifulSoup is generally **more reliable** for parsing HTML because it understands the structure of the document.

Regex can sometimes work for simple cases, but HTML can contain many edge cases that make regex extraction unreliable.

This exercise helps illustrate:

- How HTML scraping works
- The difference between **structured parsing** and **pattern matching**
- Why specialized parsing libraries are often preferred.

````python
import requests
from bs4 import BeautifulSoup
import re
import pandas as pd
from urllib.parse import urljoin

# website to scrape
url = "https://geekscoders.com"

# download webpage
response = requests.get(url)
html = response.text


# -------------------------
# Method 1: BeautifulSoup
# -------------------------

soup = BeautifulSoup(html, "html.parser")

# extract href values
bs_links = [link["href"] for link in soup.find_all("a", href=True)]

# convert to full links
bs_links_full = [urljoin(url, link) for link in bs_links]


# -------------------------
# Method 2: Regex
# -------------------------

pattern = r'<a[^>]*href="([^"]*)"'

regex_links = re.findall(pattern, html)

# convert to full links
regex_links_full = [urljoin(url, link) for link in regex_links]


# -------------------------
# Create comparison table
# -------------------------

max_len = max(len(bs_links_full), len(regex_links_full))

bs_links_full.extend([None] * (max_len - len(bs_links_full)))
regex_links_full.extend([None] * (max_len - len(regex_links_full)))

df = pd.DataFrame({
    "BeautifulSoup Links": bs_links_full,
    "Regex Links": regex_links_full
})

# check if both methods found the same link
df["Same?"] = df["BeautifulSoup Links"] == df["Regex Links"]

df


# -------------------------
# Extra comparison
# -------------------------

# print("\nNumber of links found by BeautifulSoup:", len([x for x in bs_links if x is not None]))
# print("Number of links found by Regex:", len([x for x in regex_links if x is not None]))

# # links found by BeautifulSoup but missed by regex
# missed_by_regex = set(x for x in bs_links if x is not None) - set(x for x in regex_links if x is not None)

# print("\nLinks missed by Regex:")
# for link in missed_by_regex:
#     print(link)
````

### 3. Tabular data example

#### What the Code Does

The script performs the following steps:

1. Loads the **Titanic dataset** from `titanic.csv`.
2. Uses a **regular expression** to extract titles from passenger names.
3. Stores the extracted title in a new column called **`Title`**.
4. Displays the first rows of the dataset showing the extracted titles.
5. Creates a **summary table** counting titles by **sex**.
6. Displays all **unique titles** found in the dataset.
7. Checks if any passenger names **did not match the regex pattern**.

#### Questions

1. What part of the regex captures the **title**?
2. Why is the pattern written as a **raw string (`r""`)**?
3. Modify the regex so it also allows **titles with more complex forms**, such as:
   - `Mlle`
   - `Mme`
   - `Dr`
4. How many **different titles** appear in the dataset?
5. Which titles appear **mostly for males vs females**?


````python
# import pandas so we can work with tables / csv files
import pandas as pd

# load the Titanic dataset from a csv file
titanic = pd.read_csv("titanic.csv")
titanic
````

````python
# import pandas so we can work with tables / csv files
import pandas as pd

# load the Titanic dataset from a csv file
titanic = pd.read_csv("titanic.csv")

# -----------------------------
# WHAT DOES THE REGEX MEAN?
# -----------------------------
# r",\s*([A-Za-z]+)\."
#
# r""           -> raw string, so Python does not complain about backslashes
# ,             -> find a comma
# \s*           -> find 0 or more spaces after the comma
# ([A-Za-z]+)   -> capture 1 or more letters (this is the title)
# \.            -> find a real dot
#
# example name:
# "Braund, Mr. Owen Harris"
#
# regex matches:
# ", Mr."
#
# captured part:
# "Mr"


# create a new column called Title
# .str.extract() uses regex to pull out part of the text
titanic["Title"] = titanic["name"].str.extract(r",\s*([A-Za-z]+)\.")

# show only the name and extracted title columns
print("First 10 rows with extracted title:\n")
print(titanic[["name", "Title"]].head(10))


# -----------------------------
# COUNT TITLES BY SEX
# -----------------------------
# pd.crosstab() makes a summary table
# rows = Title
# columns = sex
# values = counts
print("\nCount of titles by sex:\n")
title_table = pd.crosstab(titanic["Title"], titanic["sex"])
print(title_table)


# -----------------------------
# OPTIONAL: SEE UNIQUE TITLES
# -----------------------------
# this shows all different titles found in the dataset
print("\nUnique titles found:\n")
print(sorted(titanic["Title"].dropna().unique()))


# -----------------------------
# OPTIONAL: HOW MANY MISSING?
# -----------------------------
# check if any names did not match the regex
print("\nNumber of missing titles:")
print(titanic["Title"].isna().sum())
````
