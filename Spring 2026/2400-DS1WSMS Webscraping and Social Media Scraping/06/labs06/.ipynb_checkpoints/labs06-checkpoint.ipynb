```python
import pandas as pd # For standard data processing
import json # To handle .json outputs
import scrapy # The framework under consideration
from scrapy.crawler import CrawlerProcess # CrawlerProcess is used to run the spider
import subprocess # To work around the issue with the Reactor (see below)
import time # To measure time of computation
```

### 1. Scrapy framework

Scrapy is a framework for webscraping. It enforces **proceeding in a strictly defined manner**, which makes this approach much less user-friendly than BeautifulSoup and Selenium. In particular, it requires familiarity with object-oriented programming.

In Scrapy, you can't simply run `process.crawl()` without defining **a Spider class** because **Scrapy needs a structure that specifies how the crawling process should work**, how URLs should be selected, and how to extract data from the HTTP response.

### 1.1. Scrapy framework in Jupyter Notebook

In general Scrapy should be used outside the Jupyter Notebook. Let us demonstrate why and how to work this around.

```python
# # Define the URL to scrape
# url = 'http://quotes.toscrape.com' # This is the webpage containing quotes we want to scrape

# # Create an empty list to store the results of the crawl (quotes and authors)
# results = []

# # Create a Scrapy process to manage the crawling
# process = CrawlerProcess(settings={                 # Settings define the behavior of the Scrapy crawl
#     'LOG_LEVEL': 'ERROR',                           # Set the logging level to 'ERROR' to avoid unnecessary log messages
#     'FEEDS': {
#         'outputs/scrapy_output_jupyter.json': {     # Specify the location where the output will be saved
#             'format': 'json',                       # Define the output format to be JSON
#             'overwrite': True,                      # Overwrite if neccessary
#         },
#     },
# })

# # Define the spider class that will handle the crawling; it inherits from scrapy.Spider
# class SimpleSpider(scrapy.Spider):
#     name = 'simple'     # The name of the spider (can be used to refer to the spider)
#     start_urls = [url]  # The list of URLs to begin the crawling process. Here it's just one URL

#     # The parse method is where we define how to extract information from the webpage
#     # The parse method is automatically called by Scrapy when it loads a page
#     def parse(self, response):                  # The 'response' is the HTML content returned from the page
#         for quote in response.css('div.quote'): # Loop through each 'quote' element on the page
            
#             # Extract the quote text and author, and append them to the results list
#             yield {  # Scrapy automatically handles returning items (instead of appending to a list)
#                 'text': quote.css('span.text::text').get(),     # Extract the quote text using CSS selectors
#                 'author': quote.css('small.author::text').get() # Extract the author using CSS selectors
#             }

# # Add the SimpleSpider to the Scrapy process and start the crawl
# process.crawl(SimpleSpider) # Add the spider to the process to be run
# process.start()             # Start the crawling process
```

**CAUTION!** This does not work due to how Scrapy interacts with the Twisted reactor!

Scrapy is built on top of Twisted, a networking engine that uses an event-driven programming model. Twisted’s reactor is responsible for handling asynchronous operations and scheduling events. **The reactor can only be started once per Python process**. This means that once the reactor has been started, it cannot be restarted in the same session.

When you call `process.start()`, Scrapy starts the reactor, runs the spider, and then shuts down. However, in Jupyter, if you try to run Scrapy again (by calling `process.start()` another time), you'll get the `ReactorNotRestartable` error because the reactor is already running. This is a limitation specific to running Scrapy within an interactive environment like Jupyter Notebook, where the same Python process is being reused across multiple cells.

To solve this, we need to use a `subprocess` to **run the Scrapy spider independently**, allowing us to run it multiple times without encountering reactor issues. Here’s the updated solution where we can run the Scrapy spider multiple times from within Jupyter.

In the `subprocess` approach, we run the spider in a **completely separate process from the Jupyter notebook**. This means: the reactor is initialized and run within the subprocess; the subprocess is completely isolated from the main Jupyter session, so once the spider finishes, there is no conflict with the reactor in the notebook. You can run the spider multiple times, as each run happens in a fresh subprocess with its own reactor.

The first step is to create a `.py` file with our scraper: 

```python
scraper_code = '''
import scrapy
from scrapy.crawler import CrawlerProcess

# Define the URL to scrape
url = 'http://quotes.toscrape.com' # This is the webpage containing quotes we want to scrape

# Create an empty list to store the results of the crawl (quotes and authors)
results = []

# Create a Scrapy process to manage the crawling
process = CrawlerProcess(settings={                 # Settings define the behavior of the Scrapy crawl
    'LOG_LEVEL': 'ERROR',                           # Set the logging level to 'ERROR' to avoid unnecessary log messages
    'FEEDS': {
        'outputs/scrapy_output_jupyter_subprocess.json': {     # Specify the location where the output will be saved
            'format': 'json',                       # Define the output format to be JSON
            'overwrite': True,                      # Overwrite if neccessary
        },
    },
})

# Define the spider class that will handle the crawling; it inherits from scrapy.Spider
class SimpleSpider(scrapy.Spider):
    name = 'simple'     # The name of the spider (can be used to refer to the spider)
    start_urls = [url]  # The list of URLs to begin the crawling process. Here it's just one URL

    # The parse method is where we define how to extract information from the webpage
    # The parse method is automatically called by Scrapy when it loads a page
    def parse(self, response):                  # The 'response' is the HTML content returned from the page
        for quote in response.css('div.quote'): # Loop through each 'quote' element on the page
            
            # Extract the quote text and author, and append them to the results list
            yield {  # Scrapy automatically handles returning items (instead of appending to a list)
                'text': quote.css('span.text::text').get(),     # Extract the quote text using CSS selectors
                'author': quote.css('small.author::text').get() # Extract the author using CSS selectors
            }

# Add the SimpleSpider to the Scrapy process and start the crawl
process.crawl(SimpleSpider) # Add the spider to the process to be run
process.start()             # Start the crawling process
'''

with open("scripts/scraper.py", "w", encoding="utf-8") as file:
    file.write(scraper_code)
```

Then, one can run it from Jupyter as a subprocess:

```python
script_path = r'scripts\scraper.py'

result = subprocess.run(['python', script_path], capture_output=True, text=True, shell=True)

result
```

```python
# After the crawl finishes, check the results
with open("outputs/scrapy_output_jupyter_subprocess.json", 'r') as f:
    data = json.load(f)

pd.DataFrame(data)
```

Notably, one could also run the `scraper.py` form the command line.

### 1.2. Scrapy framework vs. BeautifulSoup and Selenium

Let us compare the time of computation between Scrapy, BeautifulSoup and Selenium.

For Scrapy we have already built the spider, so let us simply measure its computation time:

```python
start_time = time.time()

script_path = r'scripts\scraper.py'
result = subprocess.run(['python', script_path], capture_output=True, text=True, shell=True)

end_time = time.time()
computation_time = end_time - start_time
print(f"Time taken to scrape the data with Scrapy: {computation_time} seconds")
```

Were we to scrap the same page with BeautifulSoup, it would look like this:

```python
import requests
from bs4 import BeautifulSoup

start_time = time.time()

# Define the URL to scrape
url = 'http://quotes.toscrape.com'

# Send a GET request to fetch the page content
response = requests.get(url)

# Parse the content of the page using BeautifulSoup
soup = BeautifulSoup(response.text, 'html.parser')

# Create an empty list to store the results of the scrape (quotes and authors)
results = []

# Loop through each quote on the page
for quote in soup.find_all('div', class_='quote'):
    # Extract the quote text and author
    text = quote.find('span', class_='text').get_text()
    author = quote.find('small', class_='author').get_text()

    # Append the extracted data to the results list
    results.append({
        'text': text,
        'author': author
    })

# Output the results to a JSON file
with open('outputs/beautifulsoup_output.json', 'w', encoding='utf-8') as f:
    json.dump(results, f, ensure_ascii=False, indent=4)

end_time = time.time()
computation_time = end_time - start_time
print(f"Time taken to scrape the data with BeautifulSoup: {computation_time} seconds")
```

Were we to proceed with Selenium, it would look as follows:

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

path = ChromeDriverManager().install()

start_time = time.time()

service = Service(executable_path = path) 
options = webdriver.ChromeOptions()
driver = webdriver.Chrome(service = service, options = options)

# Define the URL to scrape
url = 'http://quotes.toscrape.com'

# Navigate to the page using Selenium
driver.get(url)

# Create an empty list to store the results of the scrape (quotes and authors)
results = []

# Wait for the page to load (you can use WebDriverWait if you want to ensure specific elements are loaded)
driver.implicitly_wait(10)  # Wait for elements to load, set an implicit wait

# Loop through each quote on the page
quotes = driver.find_elements(By.CSS_SELECTOR, 'div.quote')

for quote in quotes:
    # Extract the quote text and author
    text = quote.find_element(By.CSS_SELECTOR, 'span.text').text
    author = quote.find_element(By.CSS_SELECTOR, 'small.author').text

    # Append the extracted data to the results list
    results.append({
        'text': text,
        'author': author
    })

# Output the results to a JSON file
with open('outputs/selenium_output.json', 'w', encoding='utf-8') as f:
    json.dump(results, f, ensure_ascii=False, indent=4)

# Close the WebDriver
driver.quit()

end_time = time.time()
computation_time = end_time - start_time
print(f"Time taken to scrape the data with Selenium: {computation_time} seconds")
```

Selenium is undoubtedly the slowest, yet the only one to handle dynamic websites.

However, it should be emphasised that Scrapy is designed to handle **large-scale scraping projects**. It uses **asynchronous requests**, meaning it can **crawl multiple pages at once**, making it much faster when dealing with a large number of pages. Let us compare Scrapy with BeautifulSoup for 100 pages. Firstly, we need to modify the Scrapy spider:

```python
scraper_code = '''
import scrapy
from scrapy.crawler import CrawlerProcess

# Define the URL to scrape
url = 'http://quotes.toscrape.com' # This is the webpage containing quotes we want to scrape

# Create an empty list to store the results of the crawl (quotes and authors)
results = []

# Create a Scrapy process to manage the crawling
process = CrawlerProcess(settings={                 # Settings define the behavior of the Scrapy crawl
    'LOG_LEVEL': 'ERROR',                           # Set the logging level to 'ERROR' to avoid unnecessary log messages
    'FEEDS': {
        'outputs/scrapy_output_jupyter_subprocess_manypages.json': {     # Specify the location where the output will be saved
            'format': 'json',                       # Define the output format to be JSON
            'overwrite': True,                      # Overwrite if neccessary
        },
    },
})

# Define the spider class that will handle the crawling; it inherits from scrapy.Spider
class SimpleSpider(scrapy.Spider):
    name = 'simple'     # The name of the spider (can be used to refer to the spider)
    start_urls = [url]*100  # The list of URLs to begin the crawling process. Here it's 100 URLs

    # The parse method is where we define how to extract information from the webpage
    # The parse method is automatically called by Scrapy when it loads a page
    def parse(self, response):                  # The 'response' is the HTML content returned from the page
        for quote in response.css('div.quote'): # Loop through each 'quote' element on the page
            
            # Extract the quote text and author, and append them to the results list
            yield {  # Scrapy automatically handles returning items (instead of appending to a list)
                'text': quote.css('span.text::text').get(),     # Extract the quote text using CSS selectors
                'author': quote.css('small.author::text').get() # Extract the author using CSS selectors
            }

# Add the SimpleSpider to the Scrapy process and start the crawl
process.crawl(SimpleSpider) # Add the spider to the process to be run
process.start()             # Start the crawling process
'''

with open("scripts/scraper_manypages.py", "w", encoding="utf-8") as file:
    file.write(scraper_code)

start_time = time.time()

script_path = r'C:\Users\macie\OneDrive\Pulpit\Webscraping and Social Media Scraping 2025\scrapy\scripts\scraper_manypages.py'
result = subprocess.run(['python', script_path], capture_output=True, text=True, shell=True)

end_time = time.time()
computation_time = end_time - start_time
print(f"Time taken to scrape the data with Scrapy: {computation_time} seconds")
```

And here we go with BeautifulSoup:

```python
start_time = time.time()

# Define the URL to scrape
url = 'http://quotes.toscrape.com'

# Create an empty list to store all results from the 100 requests
all_results = []

# Loop to scrape the website 100 times
for i in range(100):
    # Send a GET request to fetch the page content
    response = requests.get(url)

    # Parse the content of the page using BeautifulSoup
    soup = BeautifulSoup(response.text, 'html.parser')

    # Create an empty list to store the results of this particular scrape (quotes and authors)
    results = []

    # Loop through each quote on the page
    for quote in soup.find_all('div', class_='quote'):
        # Extract the quote text and author
        text = quote.find('span', class_='text').get_text()
        author = quote.find('small', class_='author').get_text()

        # Append the extracted data to the results list
        results.append({
            'text': text,
            'author': author
        })

    # Append the results of this iteration to the overall list
    all_results.extend(results)

# Output the results to a JSON file
with open('outputs/beautifulsoup_output_manypages.json', 'w', encoding='utf-8') as f:
    json.dump(all_results, f, ensure_ascii=False, indent=4)

end_time = time.time()
computation_time = end_time - start_time
print(f"Time taken to scrape the data with BeautifulSoup: {computation_time} seconds")
```

Now the difference appears massive! Clearly, in big projects Scrapy outperforms BeautifulSoup and Selenium when it comes to the time of computation.

Above the differences in the time of computation, it should be mentioned that, unlike BeautifulSoup, Scrapy has plenty of built-in features and pipelines that make it operate efficiently.

Still, for scraping dynamic websites that load content with JavaScript, Selenium is needed. Scrapy doesn't handle JavaScript-rendered content by default. However, you can use it with other tools like Splash or Selenium to handle dynamic pages, but this makes Scrapy slower and more complex for such cases.

**In conclusion**:

- Scrapy is ideal for high-performance, **large-scale scraping projects, especially when dealing with static pages** as it offers built-in features for crawling, data extraction, and storing, making it much more efficient for large data extractions than both BeautifulSoup and Selenium;
- BeautifulSoup is better suited for **small-scale, simple scraping tasks** where you don’t need to handle many requests at once;
- Selenium is excellent **for websites with dynamic content** (JavaScript-rendered), but it’s **much slower and resource-heavy** compared to Scrapy.
