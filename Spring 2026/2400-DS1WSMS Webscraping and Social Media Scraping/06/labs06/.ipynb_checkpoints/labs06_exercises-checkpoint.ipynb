Repeat the above with Playwright and Pyppeteer (Selenium alternatives) to compare their time efficiency with Scrapy, BeautifulSoup, and Selenium.

```python
# !pip install playwright pyppeteer
# !playwright install chromium
```

Let us start with Playwright. It is a modern browser automation library by Microsoft, similar to Selenium but using the Chrome DevTools Protocol instead of WebDriver:

```python
from playwright.sync_api import sync_playwright

start_time = time.time()

# Define the URL to scrape
url = 'http://quotes.toscrape.com'

# Launch a headless Chromium browser using Playwright
with sync_playwright() as p:
    browser = p.chromium.launch(headless=True)
    page = browser.new_page()

    # Navigate to the page
    page.goto(url)

    # Wait for the page to load
    page.wait_for_selector('div.quote')

    # Create an empty list to store the results of the scrape (quotes and authors)
    results = []

    # Loop through each quote on the page
    quotes = page.query_selector_all('div.quote')

    for quote in quotes:
        # Extract the quote text and author
        text = quote.query_selector('span.text').inner_text()
        author = quote.query_selector('small.author').inner_text()

        # Append the extracted data to the results list
        results.append({
            'text': text,
            'author': author
        })

    # Close the browser
    browser.close()

# Output the results to a JSON file
with open('outputs/playwright_output.json', 'w', encoding='utf-8') as f:
    json.dump(results, f, ensure_ascii=False, indent=4)

end_time = time.time()
computation_time = end_time - start_time
print(f"Time taken to scrape the data with Playwright: {computation_time} seconds")
```

Now let us proceed with Pyppeteer. It is a Python port of Puppeteer (Node.js library by Google), also using the DevTools Protocol. Since it uses asyncio, we run it as a subprocess (same approach as with Scrapy):

```python
scraper_code = '''
import asyncio
import json
from pyppeteer import launch

# Define the URL to scrape
url = "http://quotes.toscrape.com"

async def scrape():
    # Launch a headless Chromium browser using Pyppeteer
    browser = await launch(headless=True, args=["--no-sandbox"])
    page = await browser.newPage()

    # Navigate to the page
    await page.goto(url)

    # Wait for the page to load
    await page.waitForSelector("div.quote")

    # Extract quotes using JavaScript evaluation in the browser
    results = await page.evaluate("""() => {
        const quotes = document.querySelectorAll('div.quote');
        return Array.from(quotes).map(quote => ({
            text: quote.querySelector('span.text').innerText,
            author: quote.querySelector('small.author').innerText
        }));
    }""")

    # Close the browser
    await browser.close()
    return results

# Run the async scraper
results = asyncio.get_event_loop().run_until_complete(scrape())

# Output the results to a JSON file
with open("outputs/pyppeteer_output.json", "w", encoding="utf-8") as f:
    json.dump(results, f, ensure_ascii=False, indent=4)
'''

with open("scripts/pyppeteer_scraper.py", "w", encoding="utf-8") as file:
    file.write(scraper_code)
```

```python
start_time = time.time()

script_path = r'scripts\pyppeteer_scraper.py'
result = subprocess.run(['python', script_path], capture_output=True, text=True, shell=True)

end_time = time.time()
computation_time = end_time - start_time
print(f"Time taken to scrape the data with Pyppeteer: {computation_time} seconds")
```

Now let us compare at scale. Playwright for 100 pages:

```python
start_time = time.time()

# Define the URL to scrape
url = 'http://quotes.toscrape.com'

# Create an empty list to store all results from the 100 requests
all_results = []

with sync_playwright() as p:
    browser = p.chromium.launch(headless=True)
    page = browser.new_page()

    # Loop to scrape the website 100 times
    for i in range(100):
        page.goto(url)
        page.wait_for_selector('div.quote')

        quotes = page.query_selector_all('div.quote')

        for quote in quotes:
            text = quote.query_selector('span.text').inner_text()
            author = quote.query_selector('small.author').inner_text()

            all_results.append({
                'text': text,
                'author': author
            })

    browser.close()

# Output the results to a JSON file
with open('outputs/playwright_output_manypages.json', 'w', encoding='utf-8') as f:
    json.dump(all_results, f, ensure_ascii=False, indent=4)

end_time = time.time()
computation_time = end_time - start_time
print(f"Time taken to scrape the data with Playwright: {computation_time} seconds")
```

And Pyppeteer for 100 pages:

```python
scraper_code = '''
import asyncio
import json
from pyppeteer import launch

url = "http://quotes.toscrape.com"

async def scrape_manypages():
    browser = await launch(headless=True, args=["--no-sandbox"])
    page = await browser.newPage()

    all_results = []

    # Loop to scrape the website 100 times
    for i in range(100):
        await page.goto(url)
        await page.waitForSelector("div.quote")

        results = await page.evaluate("""() => {
            const quotes = document.querySelectorAll('div.quote');
            return Array.from(quotes).map(quote => ({
                text: quote.querySelector('span.text').innerText,
                author: quote.querySelector('small.author').innerText
            }));
        }""")
        all_results.extend(results)

    await browser.close()
    return all_results

results = asyncio.get_event_loop().run_until_complete(scrape_manypages())

with open("outputs/pyppeteer_output_manypages.json", "w", encoding="utf-8") as f:
    json.dump(results, f, ensure_ascii=False, indent=4)
'''

with open("scripts/pyppeteer_scraper_manypages.py", "w", encoding="utf-8") as file:
    file.write(scraper_code)

start_time = time.time()

script_path = r'scripts\pyppeteer_scraper_manypages.py'
result = subprocess.run(['python', script_path], capture_output=True, text=True, shell=True)

end_time = time.time()
computation_time = end_time - start_time
print(f"Time taken to scrape the data with Pyppeteer: {computation_time} seconds")
```

Both Playwright and Pyppeteer should be faster than Selenium since they use the Chrome DevTools Protocol directly, avoiding the overhead of WebDriver. However, they still cannot compete with Scrapy at scale due to its asynchronous architecture.

### Exercise 2

Integrate Scrapy with Selenium and compare time efficiency with the other approaches.

```python
# !pip install scrapy-selenium
```

The `scrapy-selenium` package provides a `SeleniumRequest` that routes requests through a Selenium WebDriver instead of Scrapy's default downloader. This allows Scrapy to handle JavaScript-rendered pages while keeping its pipeline and scheduling features.

```python
scraper_code = '''
import scrapy
from scrapy.crawler import CrawlerProcess
from scrapy_selenium import SeleniumRequest

# Define the URL to scrape
url = "http://quotes.toscrape.com"

# Create a Scrapy process to manage the crawling
process = CrawlerProcess(settings={
    "LOG_LEVEL": "ERROR",
    "FEEDS": {
        "outputs/scrapy_selenium_output.json": {
            "format": "json",
            "overwrite": True,
        },
    },
    "DOWNLOADER_MIDDLEWARES": {
        "scrapy_selenium.SeleniumMiddleware": 800,
    },
    "SELENIUM_DRIVER_NAME": "chrome",
    "SELENIUM_DRIVER_ARGUMENTS": ["--headless", "--no-sandbox", "--disable-dev-shm-usage"],
})

class QuotesSeleniumSpider(scrapy.Spider):
    name = "quotes_selenium"

    def start_requests(self):
        yield SeleniumRequest(url=url, callback=self.parse)

    def parse(self, response):
        for quote in response.css("div.quote"):
            yield {
                "text": quote.css("span.text::text").get(),
                "author": quote.css("small.author::text").get(),
            }

process.crawl(QuotesSeleniumSpider)
process.start()
'''

with open("scripts/scrapy_selenium_scraper.py", "w", encoding="utf-8") as file:
    file.write(scraper_code)
```

```python
start_time = time.time()

script_path = r'scripts\scrapy_selenium_scraper.py'
result = subprocess.run(['python', script_path], capture_output=True, text=True, shell=True)

end_time = time.time()
computation_time = end_time - start_time
print(f"Time taken to scrape the data with Scrapy + Selenium: {computation_time} seconds")
```

And for 100 pages:

```python
scraper_code = '''
import scrapy
from scrapy.crawler import CrawlerProcess
from scrapy_selenium import SeleniumRequest

# Define the URL to scrape
url = "http://quotes.toscrape.com"

# Create a Scrapy process to manage the crawling
process = CrawlerProcess(settings={
    "LOG_LEVEL": "ERROR",
    "FEEDS": {
        "outputs/scrapy_selenium_output_manypages.json": {
            "format": "json",
            "overwrite": True,
        },
    },
    "DOWNLOADER_MIDDLEWARES": {
        "scrapy_selenium.SeleniumMiddleware": 800,
    },
    "SELENIUM_DRIVER_NAME": "chrome",
    "SELENIUM_DRIVER_ARGUMENTS": ["--headless", "--no-sandbox", "--disable-dev-shm-usage"],
})

class QuotesSeleniumSpider(scrapy.Spider):
    name = "quotes_selenium"

    def start_requests(self):
        # Issue 100 SeleniumRequests
        for i in range(100):
            yield SeleniumRequest(
                url=url,
                callback=self.parse,
                dont_filter=True  # Allow duplicate URLs
            )

    def parse(self, response):
        for quote in response.css("div.quote"):
            yield {
                "text": quote.css("span.text::text").get(),
                "author": quote.css("small.author::text").get(),
            }

process.crawl(QuotesSeleniumSpider)
process.start()
'''

with open("scripts/scrapy_selenium_scraper_manypages.py", "w", encoding="utf-8") as file:
    file.write(scraper_code)

start_time = time.time()

script_path = r'scripts\scrapy_selenium_scraper_manypages.py'
result = subprocess.run(['python', script_path], capture_output=True, text=True, shell=True)

end_time = time.time()
computation_time = end_time - start_time
print(f"Time taken to scrape the data with Scrapy + Selenium: {computation_time} seconds")
```

Scrapy + Selenium is the slowest approach at scale, since every request goes through a full browser render. However, it is useful when one needs Scrapy's built-in features (pipelines, middleware, scheduling) on JavaScript-rendered pages.
